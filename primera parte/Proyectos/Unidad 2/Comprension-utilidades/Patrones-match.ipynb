# Pattern Matching en Python (match-case)

---
El Pattern Matching (introducido en Python 3.10 vía PEP 634) no solo evalúa si una expresión es igual a un valor. Su verdadera potencia reside en la desestructuración: la capacidad de validar la forma (estructura) de un objeto y extraer sus partes internas en variables locales en un solo paso.

## Match tradicional

```python
hora=int(input("Ingresa la hora: "))
match hora:
    case 8:
            print ('Hora de desayunar')
    case 14:
            print ('Hora de almorzar')
    case 21:
            print ('Hora de cenar')    
    case _:
            print ('Estás a dieta!!!')
```

## Podemos tener en un case operadores booleanos

```python
mes=int(input("Ingresa el mes para ver la estación: "))
match mes:
    case 12 | 1 | 2 :
            print ('Verano')
    case 3 | 4 | 5 :
            print ('Otoño')
    case 6 | 7 | 8 :
            print ('Invierno')
    case 9 | 10 | 11 :
            print ('Primavera')    
    case _:
            print ('Error')
```

**Una de las funciones más potentes es añadir un if al case. Esto permite filtrar no solo por la "forma" de los datos, sino por su contenido.**

```python
# Ejemplo: Filtrar ventas grandes
#venta = {'tipo': 'venta', 'monto': 1500}
venta = {'tipo': 'venta', 'monto': 500}
match venta:
    case {'tipo': 'venta', 'monto': m} if m > 1000:
        print(f"Venta de alto valor: ${m} (Requiere aprobación)")
    case {'tipo': 'venta', 'monto': m}:
        print(f"Venta estándar: ${m}")
```
