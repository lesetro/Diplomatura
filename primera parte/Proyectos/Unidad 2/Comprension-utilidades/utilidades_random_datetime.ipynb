## Módulo random

Permite generar valores aleatorios útiles en simulaciones, juegos, muestreo de datos, entre otros.


### Funciones básicas

```python
import random
list
# Número entero aleatorio
print([random.randint(1, 100) for i in range(10)])
print([random.randint(1, 6) for i in range(20)])
```

```python
# Número decimal aleatorio entre 0 y 1
print(random.random())
```

```python
# Elegir un elemento al azar
colores = ['rojo', 'azul', 'verde']
print(random.choice(colores))
```

```python
# Mezclar una lista
numeros = [1,2,3,4,5]
random.shuffle(numeros)
print(numeros)
```

```python
#vamos a elegir tres estudiantes al azar de una lista de estudiantes
estudiantes = ['Ana', 'Luis', 'Pedro', 'Juan', 'Maria']
seleccionados = random.sample(estudiantes, 3)
print("Estudiantes seleccionados:", seleccionados)
# Selección de múltiples elementos
print(random.sample(numeros, 3))
```

## Módulo datetime

Permite trabajar con fechas y horas.


```python
from datetime import datetime

# Fecha y hora actual
ahora = datetime.now()
print(ahora)
```

```python
# Crear una fecha específica
fecha = datetime(2020, 4, 15)
print(f"Fecha y hora: {fecha}")
print(f"Solo fecha: {fecha.date()}")
print(f"Solo dia: {fecha.date().day}")
print(f"Solo mes: {fecha.date().month}")
print(f"Solo año: {fecha.date().year}")
```

```python
# Formateo de fechas
print(f"Fecha formateada a texto {ahora.strftime('%d/%m/%Y')}")
print(f"Hora formateada  a texto {ahora.strftime('%H:%M:%S')}")
```

```python
texto = "18/03/2026 15:30"
fecha = datetime.strptime(texto, "%d/%m/%Y %H:%M")
print(f"De texto a objeto: {fecha}")
```

```python
# Diferencia entre fechas
from datetime import timedelta

hoy = datetime.now()
print(f"Fecha actual: {hoy.date()}")
futuro = hoy + timedelta(days=10)
print(futuro)
print(futuro - hoy)
```

## Modulo timedelta
timedelta es una clase del módulo datetime que representa una duración de tiempo (no una fecha).

```python
#fechas aleatorias
from datetime import datetime, timedelta
import random

inicio = datetime(2026, 1, 1)
fin = datetime(2026, 12, 31)

dias = (fin - inicio).days
#timedelta(days=...) → crea una duración de x días
fecha_random = inicio + timedelta(days=random.randint(0, dias))
print(fecha_random)
```

```python
# fecha futura en una semana
una_semana=timedelta(weeks=1)
ahora=datetime.now()
futuro=ahora+una_semana
print(f"Fecha dentro de una semana {futuro.date().strftime('%d/%m/%Y')} ")
```
