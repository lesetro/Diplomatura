# Comprensiones en Python

Las **comprensiones** permiten crear estructuras de datos de forma concisa y legible.

Tipos principales:
- Comprensión de listas
- Comprensión de conjuntos
- Comprensión de diccionarios


## 1. Comprensión de listas

Sintaxis:
```python
[expresion for elemento in iterable if condicion]

[expresion_true if condicion else expresion_false for elemento in iterable]
```


````python
# Ejemplo básico
numeros = [1, 2, 3, 4, 5]
cuadrados = [n**2 for n in numeros]
print(cuadrados)
````

````python
# Con condición
pares = [n for n in numeros if n % 2 == 0]
print(pares)
````

## 2. Comprensión de conjuntos

Sintaxis:
```python
{expresion for elemento in iterable}
```

Los conjuntos eliminan automáticamente duplicados.

````python
# Ejemplo
numeros = [1, 2, 2, 3, 4, 4, 5]
unicos = {n for n in numeros}
print(unicos)
````

````python
# Con condición
mayores_a_dos = {n for n in numeros if n > 2}
print(mayores_a_dos)
````

````python
numeros = range(1,11)
clasifica = [
    "Múltiplo de 3" if n % 3 == 0
    else "Par" if n % 2 == 0
    else "Impar"
    for n in numeros
]
print(list(numeros))
print(clasifica)
````

````python
numeros = [5, -3, 0, 8, -1]
clasifica = [
    "Positivo" if n > 0
    else "Negativo" if n < 0
    else "Cero"
    for n in numeros
]
print(clasifica)
````

## 3. Comprensión de diccionarios

Sintaxis:
```python
{clave: valor for elemento in iterable}

{clave: valor_if_true if condicion else valor_if_false}
```

````python
# Ejemplo
numeros = [1, 2, 3, 4]
cuadrados_dict = {n: n**2 for n in numeros}
print(cuadrados_dict)
````

````python
# Con condición
pares_dict = {n: n**2 for n in numeros if n % 2 == 0}
print(pares_dict)
````

````python
numeros = range(1, 6)
clasificacion = {
    n: "Par" if n % 2 == 0 else "Impar"
    for n in numeros
}
print(clasificacion)

# otro ejemplo para tener en cuenta cuando trabajemos con diccionarios
# Comprensión de DICCIONARIO (clave: valor)
precios = [1200, 25, 350, 80, 1100]

# Crear diccionario {precio: etiqueta}
diccionario = {p: ("Caro" if p > 500 else "Barato") for p in precios}
print(diccionario)
# por lo visto en diccionarios trabaja con el : para dar clave y valor, 
# entonces el if else se puede usar para asignar un valor a cada clave 
# dependiendo de la condicion dada, en este caso si el precio es mayor 
# a 500 se asigna "Caro" y sino se asigna "Barato"
# {1200: 'Caro', 25: 'Barato', 350: 'Barato', 80: 'Barato', 1100: 'Caro'}
````

### Desafío 5
Crear un diccionario usando comprensión donde:
```python
nombres = ["Ana", "Luis", "Pedro", "Juan"]
edades = [20, 17, 25, 19]
```
-La clave sea el nombre
-El valor sea la edad
-Solo se incluyan las personas mayores o iguales a 18 años

Mostrar el diccionario resultante.

````python
nombres = ["Ana", "Luis", "Pedro", "Juan"]
edades = [20, 17, 25, 19]

diccionario = {nombre: edad for nombre, edad in zip(nombres, edades) if edad >= 18}
print(f"Diccionario resultante: {diccionario}")

diccioneario = {edad: nombre for nombre, edad in zip(nombres, edades) if edad >= 18}
print(f"Diccionario resultante: {diccioneario}")
````
