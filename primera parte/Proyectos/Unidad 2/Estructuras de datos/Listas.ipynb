## 1. Listas como Series de Datos (Vectores)
En ciencia de datos muchas veces trabajamos con series de observaciones.
Ejemplo: temperaturas registradas durante el día. Cada valor representa una observación medida.
Esto es muy común cuando se analizan:
* sensores
* datos meteorológicos
* precios
* métricas de sistemas

```python
# Temperaturas registradas en una estación meteorológica (grados Celsius)
temperaturas = [22.5, 23.0, 21.8, 19.2, 25.4]
```

Podemos acceder a los datos usando **índices**:

```python
temperaturas[0]   # primera medición
temperaturas[-1]  # última medición
print (f"Primera medición {temperaturas[0]}, última medición {temperaturas[-1]}")
```

#### Segmentación de Datos (Slicing)
El Slicing nos permite extraer una sublista (un fragmento) de nuestros datos. Esto es fundamental en Ciencia de Datos para separar, por ejemplo, los datos de entrenamiento de los de prueba o para analizar ventanas de tiempo específicas.

La sintaxis es: lista[inicio:fin]
* inicio: Es el índice donde empezamos (incluido).
* fin: Es el índice donde terminamos (excluido).

```python
# Ventas de una semana (Lunes a Domingo)
ventas = [120, 150, 130, 170, 200, 210, 190]

# 1. Revisar los días hábiles (Lunes a Viernes -> índices 0 al 4)
dias_habiles = ventas[0:5] 
print(f"Ventas días hábiles: {dias_habiles}")

# 2. Extraer el fin de semana
# Si no ponemos fin, llega hasta el último elemento
fin_de_semana = ventas[5:]
print(f"Ventas fin de semana: {fin_de_semana}")

# 3. Los últimos 3 días
ultimos_tres = ventas[-3:]
print(f"Últimas 3 mediciones: {ultimos_tres}")
```

## 4. Acceder a datos dentro una matriz 
Para acceder a un valor usamos doble **indexación**.
Esto es similar a acceder a una celda en una tabla.

```python
# 1. Seleccionar los primeros dos usuarios (Filas 0 y 1)
# El índice 2 es el límite, no se incluye.
primeros_dos = usuarios[0:2]
print(f"Primeros registros: {primeros_dos}")

# 2. Seleccionar desde el tercer usuario hasta el final
# Útil para separar una parte del dataset
desde_tercero = usuarios[2:]
print(f"Desde el índice 2 en adelante: {desde_tercero}")

# 3. Seleccionar el último usuario en una sublista
# Nota: usuarios[-1] devolvería el elemento, usuarios[-1:] devuelve una LISTA con ese elemento.
ultimo_registro = usuarios[-1:]
print(f"Último registro (como sublista): {ultimo_registro}")
```

```python
# Acceder a la puntuación de crédito del segundo usuario
segundo_registro = usuarios[1]
# Acceder a la puntuación de crédito del segundo usuario
score_usuario = usuarios[1][1]
print (f" Puntos del segundo usuario: {score_usuario}")
```

## 6. Análisis básico con listas
Aunque todavía no usamos librerías de análisis, podemos hacer análisis simples.
Ejemplo: precios de productos.
El método **sort()** sirve para ordenar alfabéticamente (o en órden numérico ascendente) los elementos de tu lista.
El método **len()** permite conocer el largo de la lista (es decir: cuántos elementos contiene).

```python
precios = [100, 50, 200, 150]
precios.sort() # analizamos rango de precios
precios
```

```python
minimo = precios[0]
maximo = precios[-1]
cantidad_muestras = len(precios)
print (f"Precio mínimo {minimo},  Precio maximo {maximo}, cantidad de muestras: {cantidad_muestras} ")
```

## 7. Operaciones útiles en listas


#### Búsqueda y Manipulación
1. Pertenencia y Ubicación (in, index)
Antes de procesar un dato, debemos saber si existe y dónde está.

**in**: Devuelve True si un elemento está en la lista. Es vital para filtrado rápido.

**index()**: Devuelve la posición de la primera aparición de un valor.

```python
# Lista de sensores activos en una planta
sensores_activos = ["TEMP_01", "PRES_02", "HUM_03", "TEMP_02"]

# ¿Está el sensor de presión activo?
existe_presion = "PRES_02" in sensores_activos 

# ¿En qué posición está el sensor HUM_03?
posicion = sensores_activos.index("HUM_03")
```

#### Frecuencia de Datos 
En análisis descriptivo, **count()** nos ayuda a entender la distribución de categorías o la repetición de eventos.

```python
# Registro de fallas en un servidor durante el mes
fallas = ["Timeout", "None", "None", "Timeout", "Overflow", "Timeout"]

# Contar cuántas veces ocurrió un "Timeout"
total_timeouts = fallas.count("Timeout")
```

#### Validación de Condiciones
El método **all()** es una función de Python que verifica si todos los elementos de un iterable son verdaderos (o cumplen una condición). En datos, se usa para validar calidad (Data Quality).

```python
# Verificación de calidad: ¿Todos los registros tienen datos (no son 0)?
lecturas = [12.5, 10.2, 14.8, 0.0]
# Esto devuelve False porque hay un 0.0 (que Python evalúa como False)
calidad_ok = all(lecturas)
```

#### Eliminar valores

Ejemplo: eliminar un dato incorrecto.El método **remove()** permite eliminar un objeto de la lista, en cuanto le pasemos como parámetro el nombre correcto del elemento.

```python
precios.remove(200)
precios
```

#### Extracción de Datos
A diferencia de remove(), que busca por valor, **pop()** elimina un elemento por su índice y, además, te lo devuelve. Es muy útil para procesar "colas" de tareas o eliminar valores atípicos (outliers) conocidos.

```python
# Tenemos una lista de tareas de procesamiento de datos
tareas = ["Limpieza", "Normalización", "Visualización"]

# Extraer la última tarea para procesarla
tarea_actual = tareas.pop() # Elimina "Visualización" y lo guarda en la variable
tarea_actual
```

## 8. Copia por Referencia vs. Copia por Valor
Este es uno de los conceptos más importantes. En Python, las listas son objetos mutables. Si simplemente igualas una lista a otra (lista_b = lista_a), no estás creando una copia, sino un acceso directo (referencia) al mismo objeto.

```python
# Reiniciamos los datos
datos_sensor = [20.5, 21.0, 19.8]
intento_copia = datos_sensor
print(f"Datos Originales: {datos_sensor}")
intento_copia.append(999)
print(f"Original afectado: {datos_sensor}") # ¡El original también cambió!

# Vuelvo a crear el original para continuar el ejemplo
datos_sensor = [20.5, 21.0, 19.8]

# OPCIÓN A: Usando el método .copy()
copia_valor_1 = datos_sensor.copy()

# OPCIÓN B: Usando slicing [:]
copia_valor_2 = datos_sensor[:]

# Modificamos las copias
copia_valor_1.append(25.0)
copia_valor_2[0] = 100.0

print(f"Datos Originales (Seguros): {datos_sensor}")
print(f"Copia con .copy(): {copia_valor_1}")
print(f"Copia con [:]: {copia_valor_2}")
```
