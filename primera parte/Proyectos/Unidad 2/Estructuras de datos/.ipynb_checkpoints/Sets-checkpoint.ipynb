# Sets

Los **sets (conjuntos)** son una estructura de datos que almacena **elementos únicos y no ordenados**.

Son muy útiles en análisis de datos cuando queremos:

- eliminar duplicados
- comparar datasets
- encontrar coincidencias entre grupos de datos

Ejemplo de uso en ciencia de datos:

- usuarios que usan **app y web**
- clientes en **dos bases de datos**
- productos presentes en **dos tiendas**

## Crear un set

Podemos crear sets usando:

- llaves `{ }`
- la función `set()`

```python
numeros = {1, 2, 3, 4, 5}

print(numeros)
print(type(numeros))
```

## Propiedades de los sets

Los sets tienen tres características importantes:

- no permiten duplicados
- no tienen orden
- no se accede por índice

```python
datos = {1,2,2,3,3,3,4}

print(datos)
```

## Eliminar duplicados de una lista

Una aplicación común en ciencia de datos es **limpiar duplicados de un dataset**.

```python
ventas = ["laptop","mouse","mouse","teclado","monitor","laptop"]

ventas_unicas = set(ventas)

print(ventas_unicas)
```

## Agregar elementos a un set

```python
usuarios = {"Ana","Luis","Maria"}
usuarios.add("Pedro")
print(usuarios)
```

## Eliminar elementos de un set

```python
usuarios.remove("Luis")
print(usuarios)
```

# Operaciones entre sets

Las operaciones de conjuntos son muy usadas para **comparar datasets**.
1. add()
2. remove()
3. union()
4. intersection()
5. difference()
6. symmetric_difference()

```python
A = {1,2,3,4}
B = {3,4,5,6}
```

## Unión

Elementos que están en **A o en B**.

```python
A.union(B)
```

## Intersección

Elementos que están en **ambos conjuntos**.

```python
A.intersection(B)
```

## Diferencia

Elementos que están en **A pero no en B**.

```python
A.difference(B)
```

## Diferencia simétrica

Elementos que están en **uno u otro conjunto pero no en ambos**.

```python
A.symmetric_difference(B)
```

## Ejemplo

Comparar usuarios de dos plataformas.
usuarios_app = {"Ana","Luis","Maria","Pedro"}
usuarios_web = {"Maria","Pedro","Sofia","Carlos"}

```python
usuarios_app = {"Ana","Luis","Maria","Pedro"}
usuarios_web = {"Maria","Pedro","Sofia","Carlos"}

#Usuarios que usan ambas plataformas
usuarios_app.intersection(usuarios_web)
```

```python
#Usuarios totales en ambas plataformas.
usuarios_app.union(usuarios_web)
```

```python
#Usuarios que solo usan la app.
usuarios_app.difference(usuarios_web)
```
