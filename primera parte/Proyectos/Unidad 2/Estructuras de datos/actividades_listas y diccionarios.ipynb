# Actividades de práctica – Análisis de datos con listas y diccionarios

## Actividad 1 – Pedidos

```python
pedidos = [
    {'cliente':'Ana','producto':'Laptop','precio':1200},
    {'cliente':'Luis','producto':'Mouse','precio':25},
    {'cliente':'Maria','producto':'Monitor','precio':300},
    {'cliente':'Pedro','producto':'Laptop','precio':1100},
    {'cliente':'Ana','producto':'Teclado','precio':80}
]
print(pedidos)

# Tu solución aquí
Cantidad_pedidos = len(pedidos)
print("Cantidad de pedidos de Ana para Laptop:", Cantidad_pedidos)

# Mostrar solo los productos comprados por Ana
productos_ana = [pedido['producto'] for pedido in pedidos if pedido['cliente'] == 'Ana']
print("Productos comprados por Ana:", productos_ana)
# mostrar pedidos con precio mayor a 500
pedidos_caros = [pedido for pedido in pedidos if pedido['precio'] > 500]
print("Pedidos con precio mayor a 500:", pedidos_caros)

# Obtener precios y el máximo

# 1. Obtenemos la lista de precios

precio = [pedido["precio"] for pedido in pedidos]

print(f"Lista de precios: {precio}")
# 2. Encontrar el precio máximo

max_precio = max(precio)

print(f"Precio máximo: {max_precio}")
# 3. Filtrar para encontrar la descripción del producto más caro

producto_max = [pedido for pedido in pedidos if pedido["precio"] == max_precio]
# 4. Mostrar resultados

print(f"\n--- Producto más caro ---")

print(f"Producto: {producto_max[0]['producto']}")

print(f"Precio: ${producto_max[0]['precio']}")

print(f"Descripción: {producto_max}")

```

### Consignas
1. ¿Cuántos pedidos hay?
2. Mostrar productos comprados por Ana
3. Mostrar pedidos con precio mayor a 500
4. Encontrar el producto más caro
