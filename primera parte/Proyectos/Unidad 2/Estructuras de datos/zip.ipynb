# La funcion zip() 

Es una función que toma varios iterables (como dos listas o dos tuplas) y los "abrocha" (como un cierre de ropa o zipper), emparejando sus elementos según su posición.

Características clave:
Función, no contenedor: A diferencia de una lista, no puedes acceder a zip[0] directamente. Debes convertirlo a una estructura de datos (como list() o tuple() o dict()) para ver su contenido.

Lazy Evaluation: Solo genera los pares cuando se le solicita (por ejemplo, en un bucle for), lo que ahorra mucha memoria cuando trabajas con millones de datos.

Corte por el más corto: Si las listas tienen diferentes tamaños, zip se detendrá cuando la lista más corta se agote.

Cada posición de las listas se empareja:

lista1[0] con lista2[0]  
lista1[1] con lista2[1]

El resultado es un **iterador de tuplas**.

Esto es muy útil en análisis de datos cuando tenemos información separada en distintas listas, por ejemplo:

- usuarios
- puntajes
- edades
- mediciones de sensores


```python
# Datos capturados por separado
estaciones = ["Norte", "Sur", "Este"]
particulas_pm10 = [45.2, 30.8, 62.1]
calidad_aire = ["Buena", "Buena", "Moderada"]

# Creamos el iterador de tuplas
reporte_zip = zip(estaciones, particulas_pm10, calidad_aire)

# 1. Convertir a lista de tuplas (Estructura de Tabla)
# Cada tupla representa una "fila" de datos
lista_reporte = list(reporte_zip)
print(f"Lista de registros: {lista_reporte}")

# 2. Convertir a diccionario (Muy útil para búsquedas rápidas)
# Nota: Debemos volver a llamar a zip() porque los iteradores se "agotan" al usarlos
estaciones_dict = dict(zip(estaciones, particulas_pm10))
print(f"Nivel de partículas por estación: {estaciones_dict}")
```

## ¿Qué tipo de objeto devuelve zip()?

`zip()` devuelve un **objeto iterable**.

Por eso muchas veces se convierte en:

- `list(zip(...))`
- `dict(zip(...))`

```python
print(type(reporte_zip))
```

```python
print(type(estaciones_dict))
```

## ¿Qué pasa si las listas tienen diferente longitud?

`zip()` se detiene cuando termina **la lista más corta**.

Esto significa que los elementos extra de la lista más larga **se ignoran**.

```python
print(type(lista_reporte))
```

## Desempaquetado 
zip() también se puede desempaquetar (unzip) usando el operador *.

```python
# Solo me interesan las estaciones y las partículas, ignoro la calidad
estaciones, particulas, _ = zip(*lista_reporte)
print (f"Estaciones: {estaciones} - particulas: {particulas}")
```

## Iteradores Agotables 
Un objeto zip se puede recorrer una sola vez. Si intentas imprimir list(lista_zip) dos veces seguidas, la segunda vez obtendrás una lista vacía.


```python
# Definimos los datos
nombres = ["Sensor_1", "Sensor_2"]
valores = [24.5, 26.8]

# Creamos el objeto zip
iterador_sensores = zip(nombres, valores)

# PRIMER USO: Convertimos a lista
print("--- Primer intento ---")
lista_datos = list(iterador_sensores)
print(f"Lista creada: {lista_datos}")

# SEGUNDO USO: Intentamos convertirlo o recorrerlo otra vez
print("\n--- Segundo intento ---")
otra_lista = list(iterador_sensores)
print(f"Lista creada: {otra_lista}") 

print(f"Lista datos si existe: {lista_datos}")
```
