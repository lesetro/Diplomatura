## 1. LISTAS

Colección ordenada, mutable, permite duplicados

```python
# Creación
lista_vacia = []
lista_numeros = [1, 2, 3, 4, 5]
lista_mixta = [1, "hola", True, 3.14]

# Operaciones que NO modifican (consulta)
len(lista_numeros)          # 5 - cantidad de elementos
lista_numeros.index(3)      # 2 - posición del elemento 3
3 in lista_numeros          # True - ¿está el 3?
min(lista_numeros)          # 1 - valor mínimo
max(lista_numeros)          # 5 - valor máximo
sum(lista_numeros)          # 15 - suma total
lista_numeros.count(2)      # 1 - cuántas veces aparece
all([True, True, False])    # False - ¿todos True?
any([False, False, True])   # True - ¿algún True?
```

```python
# Operaciones que SÍ modifican
lista = [1, 2, 3]

lista.append(4)           # [1, 2, 3, 4] - agrega al final
lista.extend([5, 6])      # [1, 2, 3, 4, 5, 6] - agrega varios
lista.insert(2, 99)       # [1, 2, 99, 3, 4, 5, 6] - inserta en posición

lista.remove(99)          # [1, 2, 3, 4, 5, 6] - elimina primera aparición
ultimo = lista.pop()      # devuelve 6, lista queda [1, 2, 3, 4, 5]
lista.pop(0)              # devuelve 1, lista queda [2, 3, 4, 5]

lista.sort()              # [2, 3, 4, 5] - ordena ascendente
lista.reverse()           # [5, 4, 3, 2] - invierte orden

lista.clear()             # [] - vacía la lista
```

```python
# Copia de listas (¡importante!)
original = [1, 2, 3]

# Copia por REFERENCIA - comparten memoria
referencia = original
referencia[0] = 99
print(original)           # [99, 2, 3] - ¡original también cambió!

# Copia por VALOR - listas independientes
original = [1, 2, 3]
copia = original.copy()   # o list(original) o original[:]
copia[0] = 99
print(original)           # [1, 2, 3] - NO cambia
print(copia)              # [99, 2, 3]
```

## 2. TUPLAS

Colección ordenada, INMUTABLE, permite duplicados

```python
# Creación
tupla_vacia = ()
tupla_numeros = (1, 2, 3, 4)
tupla_mixta = (1, "hola", True)
tupla_un_elemento = (5,)      # ¡la coma es obligatoria!

# Acceso (igual que listas)
tupla_numeros[0]      # 1
tupla_numeros[1:3]    # (2, 3)

# Métodos disponibles (solo consulta)
tupla_numeros.count(2)   # 1
tupla_numeros.index(3)   # 2

# ❌ NO se puede modificar
# tupla_numeros[0] = 99  # ¡ERROR! TypeError
```

## 3. SETS (Conjuntos)

Colección NO ordenada, mutable, SIN duplicados

```python
# Creación
set_vacio = set()           # ¡{} es diccionario!
set_numeros = {1, 2, 3, 3, 3}  # {1, 2, 3} - elimina duplicados

# Operaciones básicas
s = {1, 2, 3}
s.add(4)                   # {1, 2, 3, 4}
s.remove(2)                # {1, 3, 4} - error si no existe
s.discard(10)              # no da error si no existe
elemento = s.pop()         # elimina y devuelve uno al azar
s.clear()                  # set() - vacía

# Operaciones entre sets
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

A.union(B)                 # {1, 2, 3, 4, 5, 6} - todos
A.intersection(B)          # {3, 4} - comunes
A.difference(B)            # {1, 2} - solo en A
B.difference(A)            # {5, 6} - solo en B
A.symmetric_difference(B)  # {1, 2, 5, 6} - no comunes

# Verificaciones
A.issubset(B)              # False - ¿A está dentro de B?
A.issuperset(B)            # False - ¿A contiene a B?
A.isdisjoint(B)            # False - ¿no tienen elementos en común?
```

## 4. ZIP

Empareja elementos de iterables por posición

```python
nombres = ["Ana", "Luis", "Carlos"]
edades = [25, 30, 28]

# Crear pares
pares = list(zip(nombres, edades))
print(pares)  # [('Ana', 25), ('Luis', 30), ('Carlos', 28)]

# Usar en loop
for nombre, edad in zip(nombres, edades):
    print(f"{nombre} tiene {edad} años")

# Con más de dos listas
ciudades = ["Buenos Aires", "Bogotá", "Lima"]
for nombre, edad, ciudad in zip(nombres, edades, ciudades):
    print(f"{nombre} ({edad}) vive en {ciudad}")
```

## 5. ENUMERATE

Agrega contador automático a un iterable

```python
colores = ["rojo", "azul", "verde"]

# Con enumerate (índice por defecto 0)
for indice, color in enumerate(colores):
    print(f"{indice}: {color}")

# Con índice personalizado
for indice, color in enumerate(colores, start=1):
    print(f"{indice}: {color}")

# Convertir a lista de tuplas
list(enumerate(colores))  # [(0, 'rojo'), (1, 'azul'), (2, 'verde')]
```

## 6. COMPREHENSION DE LISTAS

Forma compacta de crear listas

```python
# Sintaxis: [expresión for elemento in iterable if condición]

# Cuadrados de números
cuadrados = [x**2 for x in range(5)]
print(cuadrados)  # [0, 1, 4, 9, 16]

# Filtrar números pares
pares = [x for x in range(10) if x % 2 == 0]
print(pares)  # [0, 2, 4, 6, 8]

# Con if-else
etiquetas = ["par" if x % 2 == 0 else "impar" for x in range(5)]
print(etiquetas)  # ['par', 'impar', 'par', 'impar', 'par']

# Con dos iterables
pares_ordenados = [(x, y) for x in [1, 2] for y in [3, 4]]
print(pares_ordenados)  # [(1, 3), (1, 4), (2, 3), (2, 4)]
```

## 7. MATCH - CASE (Python 3.10+)

Coincidencia de patrones

```python
# Por tipo
def identificar(tipo):
    match tipo:
        case int():
            print("Es un número entero")
        case str():
            print("Es un string")
        case list():
            print("Es una lista")
        case _:
            print("Es otro tipo")

identificar(5)      # "Es un número entero"
identificar("hola") # "Es un string"
```

```python
# Por valor
def evaluar_nota(puntaje):
    match puntaje:
        case 10 | 9:
            print("Excelente")
        case 8 | 7:
            print("Muy bien")
        case 6:
            print("Aprobado")
        case _:
            print("Necesita mejorar")

evaluar_nota(9)  # "Excelente"
```

```python
# Con estructuras de datos
def procesar(dato):
    match dato:
        case {"nombre": nombre, "edad": edad}:
            print(f"Cliente: {nombre}, {edad} años")
        case [primero, segundo]:
            print(f"Lista de dos: {primero} y {segundo}")
        case _:
            print("Formato no reconocido")

procesar({"nombre": "Ana", "edad": 30})
procesar([1, 2])
```

## 8. MÓDULO RANDOM

Generación de números y selecciones aleatorias

```python
from random import *

# Números aleatorios
randint(1, 10)           # Entero entre 1 y 10 (ambos incluidos)
uniform(1.5, 5.5)        # Float entre 1.5 y 5.5
random()                 # Float entre 0.0 y 1.0

# Selección de elementos
colores = ['rojo', 'azul', 'verde', 'amarillo']
choice(colores)          # Elige uno al azar
choices(colores, k=2)    # Elige 2 al azar (con repetición)
sample(colores, k=2)     # Elige 2 al azar (sin repetición)

# Mezclar listas
numeros = [1, 2, 3, 4, 5]
shuffle(numeros)         # Mezcla la lista (la modifica)
print(numeros)           # [3, 5, 1, 4, 2] (orden aleatorio)
```

## 9. MÓDULO DATETIME

Manejo de fechas y horas

```python
from datetime import datetime, date, timedelta

# Obtener fechas actuales
ahora = datetime.now()        # Fecha y hora actual
hoy = date.today()            # Solo fecha actual

# Crear fechas específicas
fecha = date(2025, 12, 25)    # 25 de diciembre de 2025
fecha_hora = datetime(2025, 12, 25, 18, 30)  # con hora 18:30

# Extraer componentes
fecha.year      # 2025
fecha.month     # 12
fecha.day       # 25
fecha_hora.hour # 18
fecha_hora.minute # 30

# Día de la semana
fecha.weekday()     # 0=Lunes, 6=Domingo
fecha.isoweekday()  # 1=Lunes, 7=Domingo
```

```python
# Formatear fechas (datetime → string)
fecha = date(2025, 12, 25)

fecha.strftime("%d/%m/%Y")      # "25/12/2025"
fecha.strftime("%Y-%m-%d")      # "2025-12-25"
fecha.strftime("%A, %B %d")     # "Thursday, December 25"
fecha.strftime("%a, %b %d")     # "Thu, Dec 25"

# Códigos más comunes:
# %d - día (01-31)     %m - mes (01-12)     %Y - año (2025)
# %H - hora (00-23)    %M - minuto (00-59)  %S - segundo (00-59)
# %A - día semana completo   %a - día semana abreviado
# %B - mes completo           %b - mes abreviado
```

```python
# Parsear fechas (string → datetime)
texto = "25/12/2025"
fecha_parseada = datetime.strptime(texto, "%d/%m/%Y")
print(fecha_parseada)  # 2025-12-25 00:00:00

texto2 = "2025-12-25 18:30:00"
fecha2 = datetime.strptime(texto2, "%Y-%m-%d %H:%M:%S")
```

```python
# Aritmética con timedelta
hoy = datetime.now()

# Sumar tiempo
manana = hoy + timedelta(days=1)
proxima_semana = hoy + timedelta(weeks=1)
en_2_horas = hoy + timedelta(hours=2)
dentro_30_min = hoy + timedelta(minutes=30)

# Restar tiempo
ayer = hoy - timedelta(days=1)
hace_una_semana = hoy - timedelta(weeks=1)

# Diferencia entre fechas
fecha_evento = datetime(2026, 12, 31)
diferencia = fecha_evento - hoy
print(diferencia.days)        # Días de diferencia
print(diferencia.total_seconds())  # Segundos totales
```

```python
# Comparar fechas
if hoy > fecha_evento:
    print("La fecha ya pasó")
else:
    print("La fecha es en el futuro")

# Crear fechas específicas
fecha_navidad = date(2026, 12, 25)
fecha_ayer = date.today() - timedelta(days=1)
```

## 10. EJEMPLOS COMBINADOS

Aplicando múltiples conceptos juntos

```python
# Ejemplo 1: Crear lista de tuplas con enumerate y zip
nombres = ["Ana", "Luis", "Carlos"]
edades = [25, 30, 28]

resultado = [(i, nombre, edad) for i, (nombre, edad) in enumerate(zip(nombres, edades), 1)]
print(resultado)  # [(1, 'Ana', 25), (2, 'Luis', 30), (3, 'Carlos', 28)]
```

```python
# Ejemplo 2: Filtrar y transformar con comprehension
numeros = [10, 25, 32, 47, 53, 68, 71, 84, 99]

# Pares multiplicados por 2
pares_dobles = [x * 2 for x in numeros if x % 2 == 0]
print(pares_dobles)  # [20, 64, 136, 168]
```

```python
# Ejemplo 3: Combinar random, datetime y match
from random import choice
from datetime import datetime

def procesar_pedido():
    estados = ["pendiente", "procesando", "enviado", "entregado"]
    estado = choice(estados)
    fecha = datetime.now().strftime("%d/%m/%Y %H:%M")
    
    match estado:
        case "entregado":
            return f"[{fecha}] ✅ Pedido entregado"
        case "enviado":
            return f"[{fecha}] 📦 Pedido enviado"
        case "procesando":
            return f"[{fecha}] ⏳ Pedido en proceso"
        case _:
            return f"[{fecha}] ⏰ Pedido pendiente"

print(procesar_pedido())
```

```python
# Ejemplo 4: Encontrar duplicados con sets
lista_con_duplicados = [1, 2, 3, 2, 4, 5, 3, 6, 1]
duplicados = list(set([x for x in lista_con_duplicados if lista_con_duplicados.count(x) > 1]))
print(f"Duplicados: {duplicados}")  # [1, 2, 3]
```

```python
# Ejemplo 5: Calcular edad y próximos cumpleaños
from datetime import date, timedelta

def info_cumpleaños(fecha_nacimiento_str):
    # Convertir string a fecha
    nacimiento = datetime.strptime(fecha_nacimiento_str, "%d/%m/%Y").date()
    hoy = date.today()
    
    # Calcular edad
    edad = hoy.year - nacimiento.year
    if (hoy.month, hoy.day) < (nacimiento.month, nacimiento.day):
        edad -= 1
    
    # Calcular próximo cumpleaños
    prox_cumple = date(hoy.year, nacimiento.month, nacimiento.day)
    if prox_cumple < hoy:
        prox_cumple = date(hoy.year + 1, nacimiento.month, nacimiento.day)
    
    dias_faltan = (prox_cumple - hoy).days
    
    return {
        "edad": edad,
        "dias_para_cumple": dias_faltan,
        "proximo_cumple": prox_cumple.strftime("%d/%m/%Y")
    }

# Prueba
info = info_cumpleaños("15/03/1990")
print(f"Edad: {info['edad']} años")
print(f"Días para próximo cumpleaños: {info['dias_para_cumple']}")
print(f"Próximo cumpleaños: {info['proximo_cumple']}")
```

## 💡 RESUMEN RÁPIDO

| Tipo | Características | Uso típico |
|------|-----------------|------------|
| **Lista** | Ordenada, mutable, permite duplicados | Colecciones que cambian |
| **Tupla** | Ordenada, inmutable, permite duplicados | Datos fijos, claves de diccionario |
| **Set** | No ordenado, mutable, sin duplicados | Eliminar duplicados, operaciones de conjuntos |
| **zip** | Empareja iterables | Combinar listas relacionadas |
| **enumerate** | Agrega contador | Necesitar índice y valor |
| **comprehension** | Creación compacta | Transformar/filtrar listas |
| **match-case** | Coincidencia de patrones | Múltiples condiciones |
| **random** | Valores aleatorios | Sorteos, juegos, muestras |
| **datetime** | Fechas y horas | Cálculos temporales, formatos |
