# 📘 DICCIONARIOS EN PYTHON - GUÍA COMPLETA

Los diccionarios son estructuras de datos **clave-valor**. Son **mutables**, **ordenados** (desde Python 3.7) y **no permiten claves duplicadas**.

## 1. CREACIÓN DE DICCIONARIOS

```python
# Forma 1: llaves {}
dic_vacio = {}
dic_persona = {
    "nombre": "Ana",
    "edad": 30,
    "ciudad": "Buenos Aires"
}

diccionario = {
    "clave1": "valor1",
    "clave2": "valor2",
    "clave3": "valor3"
}
# Forma 2: dict()
dic_persona2 = dict(
    nombre="Luis", 
    edad=25, 
    ciudad="Lima"
    )

# Forma 3: desde listas de tuplas
pares = [
    ("nombre", "Carlos"), 
    ("edad", 28)
    ]

dic_persona3 = dict(pares)

print("Forma 1:", dic_persona)
print("Forma 2:", dic_persona2)
print("Forma 3:", dic_persona3)
```

## 2. ACCESO A VALORES

```python
persona = {
    "nombre": "Ana",
    "edad": 30,
    "ciudad": "Buenos Aires"
}

# Forma 1: con corchetes (lanza error si no existe)
print("persona['nombre']:", persona["nombre"])

# Forma 2: con get() (retorna None si no existe)
print("persona.get('nombre'):", persona.get("nombre"))
print("persona.get('telefono'):", persona.get("telefono"))
print("persona.get('telefono', 'No especificado'):", persona.get("telefono", "No especificado"))

# Obtener todas las claves
print("\npersona.keys():", list(persona.keys()))

# Obtener todos los valores
print("persona.values():", list(persona.values()))

# Obtener pares clave-valor
print("persona.items():", list(persona.items()))
```

## 3. MODIFICAR Y AGREGAR

```python
persona = {
    "nombre": "Ana",
    "edad": 30
}

# Modificar valor existente
persona["edad"] = 31
print("Modificar edad:", persona)

# Agregar nueva clave-valor
persona["ciudad"] = "Buenos Aires"
persona["telefono"] = "123456789"
print("Agregar campos:", persona)

# Actualizar múltiples valores
persona.update({
    "edad": 32, 
    "profesion": "Ingeniera"
    })
print("Actualizar múltiples:", persona)
```

## 4. ELIMINAR ELEMENTOS

```python
persona = {
    "nombre": "Ana",
    "edad": 30,
    "ciudad": "Buenos Aires",
    "telefono": "123456789"
}

# pop() - elimina y devuelve el valor
edad = persona.pop("edad")
print(f"pop('edad'): {edad}")
print("Diccionario:", persona)

# popitem() - elimina y devuelve el último par
ultimo = persona.popitem()
print(f"popitem(): {ultimo}")
print("Diccionario:", persona)

# del - elimina por clave
del persona["ciudad"]
print("después de del:", persona)

# clear() - vacía todo
persona.clear()
print("después de clear():", persona)
```

## 5. ITERAR SOBRE DICCIONARIOS

```python
persona = {
    "nombre": "Ana",
    "edad": 30,
    "ciudad": "Buenos Aires",
    "profesion": "Ingeniera"
}

print("Iterar sobre claves:")
for clave in persona:
    print(f"  {clave}")

print("\nIterar sobre valores:")
for valor in persona.values():
    print(f"  {valor}")

print("\nIterar sobre pares clave-valor:")
for clave, valor in persona.items():
    print(f"  {clave}: {valor}")
```

## 6. COMPREHENSION DE DICCIONARIOS

```python
# Crear diccionario con cuadrados
cuadrados = {x: x**2 for x in range(1, 6)}
print("Cuadrados:", cuadrados)

# Filtrar elementos
pares = {x: x**2 for x in range(1, 11) if x % 2 == 0}
print("Pares al cuadrado:", pares)

# Invertir diccionario (clave ↔ valor)
original = {"a": 1, "b": 2, "c": 3}
invertido = {valor: clave for clave, valor in original.items()}
print("Invertido:", invertido)

# Transformar valores
nombres = ["Ana", "Luis", "Carlos"]
dic_nombres = {nombre: len(nombre) for nombre in nombres}
print("Nombre -> longitud:", dic_nombres)
```

## 7. DICCIONARIOS ANIDADOS

```python
estudiantes = {
    "2024001": {
        "nombre": "Ana",
        "edad": 20,
        "cursos": ["Python", "SQL"]
    },
    "2024002": {
        "nombre": "Luis",
        "edad": 22,
        "cursos": ["Python", "Machine Learning"]
    },
    "2024003": {
        "nombre": "Carlos",
        "edad": 21,
        "cursos": ["SQL", "Data Analysis"]
    }
}

# Acceder a valores anidados
print("estudiantes['2024001']['nombre']:", estudiantes["2024001"]["nombre"])
print("estudiantes['2024002']['cursos'][0]:", estudiantes["2024002"]["cursos"][0])

# Modificar valores anidados
estudiantes["2024001"]["edad"] = 21
print("\nModificada edad:", estudiantes["2024001"])

# Agregar a lista dentro del diccionario
estudiantes["2024003"]["cursos"].append("Power BI")
print("Cursos actualizados:", estudiantes["2024003"]["cursos"])

# Iterar sobre diccionario anidado
print("\nIterar sobre estudiantes:")
for legajo, info in estudiantes.items():
    print(f"\nLegajo: {legajo}")
    for campo, valor in info.items():
        print(f"  {campo}: {valor}")
```

## 8. MÉTODOS ÚTILES

```python
# fromkeys() - crear diccionario con claves y valor por defecto
claves = ["nombre", "edad", "ciudad"]
dic_default = dict.fromkeys(claves, "desconocido")
print("fromkeys:", dic_default)

# setdefault() - obtener valor o establecer si no existe
dic = {"nombre": "Ana"}
telefono = dic.setdefault("telefono", "123456789")
print("setdefault:", telefono)
print("Dic después:", dic)

# | (merge) y |= (update) - Python 3.9+
dic1 = {"a": 1, "b": 2}
dic2 = {"b": 3, "c": 4}
merged = dic1 | dic2
print("Merge:", merged)

# Actualizar en lugar
dic1 |= dic2
print("Update in place:", dic1)
```

## 9. EJEMPLOS PRÁCTICOS

### Ejemplo 1: Contador de palabras

```python
texto = "hola mundo hola python mundo hola"
palabras = texto.split()
print("Palabras:", palabras)
contador = {}
for palabra in palabras:
    if palabra in contador:
        contador[palabra] += 1
    else:
        contador[palabra] = 1

print("Contador:", contador)
```

### Ejemplo 2: Usando defaultdict (más elegante)

```python
from collections import defaultdict

texto = "hola mundo hola python mundo hola"
palabras = texto.split()

contador = defaultdict(int)  # valor por defecto 0
for palabra in palabras:
    contador[palabra] += 1

print("Con defaultdict:", dict(contador))
```

### Ejemplo 3: Agrupar por categoría

```python
productos = [
    {"nombre": "manzana", "categoria": "fruta", "precio": 500},
    {"nombre": "pera", "categoria": "fruta", "precio": 300},
    {"nombre": "zanahoria", "categoria": "verdura", "precio": 200},
    {"nombre": "lechuga", "categoria": "verdura", "precio": 150},
]

# Agrupar por categoría
agrupados = {}
for producto in productos:
    categoria = producto["categoria"]
    if categoria not in agrupados:
        agrupados[categoria] = []
    agrupados[categoria].append(producto)

print("Agrupados por categoría:")
for cat, items in agrupados.items():
    print(f"  {cat}: {len(items)} productos")
```

### Ejemplo 4: Combinar diccionarios con zip

```python
nombres = ["Ana", "Luis", "Carlos"]
edades = [25, 30, 28]

# Crear diccionario a partir de dos listas
personas = dict(zip(nombres, edades))
print("personas:", personas)

# Crear diccionario con más estructura
personas_completo = {nombre: {"edad": edad} for nombre, edad in zip(nombres, edades)}
print("personas_completo:", personas_completo)
```

### Ejemplo 5: Diccionario como switch (alternativa a match)

```python
def operacion(simbolo, a, b):
    operaciones = {
        "+": a + b,
        "-": a - b,
        "*": a * b,
        "/": a / b if b != 0 else "Error: división por cero"
    }
    return operaciones.get(simbolo, "Operación no válida")

print("operacion('+', 5, 3):", operacion("+", 5, 3))
print("operacion('*', 5, 3):", operacion("*", 5, 3))
print("operacion('%', 5, 3):", operacion("%", 5, 3))
```

### Ejemplo 6: Ordenar diccionarios

```python
dic = {"b": 2, "a": 1, "c": 3}

# Ordenar por clave
por_clave = dict(sorted(dic.items()))
print("Ordenar por clave:", por_clave)

# Ordenar por valor
por_valor = dict(sorted(dic.items(), key=lambda item: item[1]))
print("Ordenar por valor:", por_valor)

# Orden descendente por valor
por_valor_desc = dict(sorted(dic.items(), key=lambda item: item[1], reverse=True))
print("Orden descendente por valor:", por_valor_desc)
```

### Ejercicio 1: Encontrar la persona más joven

```python
# Datos de prueba
personas = {
    "Ana": 25,
    "Luis": 30,
    "Carlos": 22,
    "Sofia": 28
}

# Solución
mas_joven = min(personas, key=personas.get)
print(f"La persona más joven es {mas_joven} con {personas[mas_joven]} años")
```

### Ejercicio 3: Contar estudiantes por curso

```python
estudiantes = {
    "Ana": ["Python", "SQL"],
    "Luis": ["Python", "Machine Learning"],
    "Carlos": ["SQL", "Data Analysis"],
    "Sofia": ["Python", "SQL", "Data Analysis"]
}

# Solución
cursos = {}
for alumno, lista_cursos in estudiantes.items():
    for curso in lista_cursos:
        cursos[curso] = cursos.get(curso, 0) + 1

print("Estudiantes por curso:", cursos)
```

### Ejercicio 4: Agenda telefónica

```python
agenda = {}

def agregar_contacto(nombre, telefono, email):
    agenda[nombre] = {"telefono": telefono, "email": email}

def buscar_contacto(nombre):
    return agenda.get(nombre, "Contacto no encontrado")

def eliminar_contacto(nombre):
    if nombre in agenda:
        del agenda[nombre]
        return f"{nombre} eliminado"
    return "Contacto no encontrado"

# Prueba
agregar_contacto("Ana", "123456789", "ana@email.com")
agregar_contacto("Luis", "987654321", "luis@email.com")

print("Buscar Ana:", buscar_contacto("Ana"))
print("Agenda:", agenda)
print(eliminar_contacto("Luis"))
print("Agenda después:", agenda)
```

## 💡 RESUMEN RÁPIDO

| Operación | Sintaxis |
|-----------|----------|
| Crear diccionario | `{}` o `dict()` |
| Acceder valor | `dic['clave']` o `dic.get('clave')` |
| Agregar/Modificar | `dic['clave'] = valor` |
| Eliminar | `del dic['clave']` o `dic.pop('clave')` |
| Ver claves | `dic.keys()` |
| Ver valores | `dic.values()` |
| Ver pares | `dic.items()` |
| Iterar | `for clave, valor in dic.items():` |
| Comprehension | `{clave: valor for item in iterable}` |
| Unir diccionarios | `dic1 \| dic2` (Python 3.9+) |
