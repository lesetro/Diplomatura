## Ejercicio 1 - Creación de arrays

1. Anatomía de un Array
Consigna: La base de NumPy es el ndarray. En este ejercicio exploraremos cómo se construyen y qué propiedades los definen.

1. Crea un array de una dimensión (1D) con 10 elementos aleatorios.
2. Crea una matriz de 3x3 (2D) rellena con ceros.
3. Utiliza los atributos de NumPy para imprimir en pantalla:
    *   La forma (shape).
    *   El tipo de datos (dtype).
    *   El número de dimensiones (ndim).
    *   La cantidad total de elementos (size).
4. Convierte el array de ceros (que por defecto es float64) a tipo de dato entero (int32).

```python
import numpy as np

# 1. Array 1D con 10 elementos aleatorios
array_1d = np.random.rand(10)
print("Array 1D aleatorio:")
print(array_1d)

# 2. Matriz 3x3 rellena con ceros
matriz_ceros = np.zeros((3, 3))
print("\nMatriz 3x3 de ceros:")
print(matriz_ceros)

# 3. Atributos de la matriz de ceros
print("\n--- Atributos de la matriz de ceros ---")
print(f"Forma (shape):        {matriz_ceros.shape}")
print(f"Tipo de datos (dtype): {matriz_ceros.dtype}")
print(f"Dimensiones (ndim):    {matriz_ceros.ndim}")
print(f"Cantidad de elementos (size): {matriz_ceros.size}")

# 4. Convertir a int32
matriz_int = matriz_ceros.astype(np.int32)
print("\nMatriz convertida a int32:")
print(matriz_int)
print(f"Nuevo dtype: {matriz_int.dtype}")
```
