## Ejercicio 1 - Anatomía de un Array

```python
import numpy as np

# 1. Array 1D con 10 elementos aleatorios
array_1d = np.random.rand(10)
print("Array 1D aleatorio:")
print(array_1d)

# 2. Matriz 3x3 rellena con ceros
matriz_ceros = np.zeros((3, 3))
print("\nMatriz 3x3 de ceros:")
print(matriz_ceros)

# 3. Atributos de la matriz de ceros
print("\n--- Atributos de la matriz de ceros ---")
print(f"Forma (shape):        {matriz_ceros.shape}")
print(f"Tipo de datos (dtype): {matriz_ceros.dtype}")
print(f"Dimensiones (ndim):    {matriz_ceros.ndim}")
print(f"Cantidad de elementos (size): {matriz_ceros.size}")

# 4. Convertir a int32
matriz_int = matriz_ceros.astype(np.int32)
print("\nMatriz convertida a int32:")
print(matriz_int)
print(f"Nuevo dtype: {matriz_int.dtype}")
```

## Ejercicio 2 - Eficiencia y Gestión de Memoria

```python
import numpy as np
import sys

# 1. Lista de Python con 1.000.000 de enteros
lista_python = list(range(1_000_000))

# 2. Array de NumPy con los mismos elementos
array_numpy = np.arange(1_000_000)

# 3. Comparación de memoria
# Para la lista: getsizeof devuelve solo el contenedor; sumamos los enteros
memoria_lista = sys.getsizeof(lista_python) + sum(sys.getsizeof(x) for x in lista_python)
memoria_array = array_numpy.nbytes

print(f"Memoria lista de Python: {memoria_lista:,} bytes ({memoria_lista/1024/1024:.2f} MB)")
print(f"Memoria array NumPy:     {memoria_array:,} bytes ({memoria_array/1024/1024:.2f} MB)")

# 4. Conclusión
veces_mas_eficiente = memoria_lista / memoria_array
print(f"\nNumPy es aproximadamente {veces_mas_eficiente:.1f} veces más eficiente en memoria.")
print("Esto se debe a que NumPy almacena los datos en un bloque contiguo de memoria")
print("con un tipo de dato fijo (int64 = 8 bytes por elemento), mientras que Python")
print("guarda cada entero como un objeto completo con metadatos (~28 bytes c/u).")
print("NumPy es más eficiente porque no desperdicia casilleros en metadatos y mantiene todo ordenado.")
```

## Ejercicio 3 - Selección Avanzada (Slicing e Indexado)

```python
import numpy as np

# 1. Matriz 5x5 con números del 1 al 25
matriz = np.arange(1, 26).reshape(5, 5)
print("Matriz original 5x5:")
print(matriz)

# 2. Tercera fila completa (índice 2)
tercera_fila = matriz[2, :]
print(f"\nTercera fila: {tercera_fila}")
tercera_fila_2 = matriz[2]
print(f"Tercera fila (otra forma): {tercera_fila_2}")

# 3. Dos últimas columnas
ultimas_columnas = matriz[:, -2:]
print("\nDos últimas columnas:")
print(ultimas_columnas)


# 4. Submatriz central de 3x3 (excluyendo bordes)
submatriz_central = matriz[1:4, 1:4]
print("\nSubmatriz central 3x3:")
print(submatriz_central)

# 5. Invertir el orden de las filas
matriz_invertida = matriz[::-1]
print("\nMatriz con filas invertidas:")
print(matriz_invertida)
```

## Ejercicio 4 - ventas_altas (Máscara Booleana)

```python
import numpy as np

ventas = np.array([120, 600, 450, 800, 300, 900, 150])

# 1. Máscara booleana para ventas > 500
mask = ventas > 500

# 2. Aplicar la máscara
ventas_altas = ventas[mask]

print(f"Ventas originales: {ventas}")
print(f"Máscara: {mask}")
print(f"Ventas filtradas (> 500): {ventas_altas}")
```

## Ejercicio 5 - Estados (np.where)

```python
import numpy as np

presiones = np.array([95, 105, 110, 80, 120, 99])

# Crear array 'estados' usando np.where
estados = np.where(presiones > 100, "ALERTA", "NORMAL")

print(f"Presiones: {presiones}")
print(f"Estados:   {estados}")
```
