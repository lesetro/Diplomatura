## Referencia rápida

Si te trabaste en alguna celda, acá tenés los conceptos clave:

| Tarea | Código |
|-------|--------|
| Crear array | `np.array([...])` |
| Ceros / Unos | `np.zeros((f,c))` / `np.ones((f,c))` |
| Secuencia | `np.arange(ini, fin, paso)` |
| Equiespaciados | `np.linspace(ini, fin, n)` |
| Forma | `arr.shape` |
| Tipo | `arr.dtype` |
| Fila i | `arr[i]` |
| Celda (i,j) | `arr[i, j]` |
| Rango de filas | `arr[a:b]` |
| Columna j | `arr[:, j]` |
| Filtrar | `arr[arr > valor]` |
| Combinar AND | `(c1) & (c2)` |
| Reclasificar | `np.where(cond, val_true, val_false)` |
| Cambiar forma | `arr.reshape(f, c)` |
| Aplanar vista | `arr.ravel()` |
| Aplanar con copia | `arr.flatten()` |
| Transponer | `arr.T` |
| Apilar filas | `np.vstack([a, b])` |
| Apilar columnas | `np.hstack([a, b])` |
| Suma / Media | `arr.sum(axis=)` / `arr.mean(axis=)` |
| Producto matricial | `A @ B` |
| Determinante | `np.linalg.det(A)` |
| Inversa | `np.linalg.inv(A)` |
| Autovalores | `np.linalg.eig(A)` |
| Normalizar | `(x - x.min()) / (x.max() - x.min())` |
| Estandarizar | `(x - x.mean()) / x.std()` |
