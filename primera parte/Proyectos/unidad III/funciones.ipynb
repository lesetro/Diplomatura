# Funciones en python

Una función es un bloque de código organizado y **reutilizable** que realiza una tarea específica.

### ¿Por qué usar funciones?
Las funciones son esenciales porque permiten:
- **Reutilizar código:** Escribe un bloque de código una vez y úsalo en múltiples lugares, evitando la repetición.
- **Evitar duplicación:** Al reutilizar código, se facilitan las actualizaciones y disminuye la probabilidad de errores.
- **Organizar lógica:** Dividen tu programa en unidades lógicas y manejables, haciendo que el código sea más legible, comprensible y fácil de corregir.
- **Dividir programas complejos:** Permiten descomponer problemas grandes en tareas más pequeñas y específicas, lo que simplifica el desarrollo y la depuración.
- **Preparar código para pipelines de datos:** En el análisis de datos, las funciones son clave para crear flujos de trabajo (pipelines) reproducibles y robustos, donde cada paso (limpieza, transformación, análisis) puede ser una función.

En esta unidad, exploraremos cómo definir, usar y optimizar funciones en Python.

### 1. Definición y Llamada de Funciones

En Python, las funciones se definen utilizando la palabra clave `def`, seguida del nombre de la función, un paréntesis que puede contener parámetros (entradas) y finalmente dos puntos (`:`).

**Sintaxis básica:**
```python
def nombre_de_la_funcion(parametro1, parametro2, ...):
    # Bloque de código de la función
    # Realiza una serie de operaciones
    return resultado # Opcional: puede devolver valores
```

La **indentación** es crucial en Python para definir el alcance del código que pertenece a la función. Una vez definida, la función puede ser "llamada" o "invocada" para ejecutar su bloque de código.

````python
#Ejemplo: definimos la funcion
def calcular_promedio(lista):
    return sum(lista) / len(lista)
````

Una vez que has definido una función, esta ya existe en el contexto de tu programa (este cuaderno), y puedes llamarla tantas veces como quieras, sin tener que escribir su código otra vez.

````python
#Ejemplo: definimos la funcion
def calcular_promedio(lista):
    return sum(lista) / len(lista)

# supongamos que debemos calcular el promedio de las notas de los alumnos
notas1 = [7,8,9]
notas2=[6,10,8]

# invocación
conjuntos_de_notas = [notas1, notas2, [5,7,7]]
for i, notas_actuales in enumerate(conjuntos_de_notas):
  promedio = calcular_promedio(notas_actuales)
  print(f"El promedio para el conjunto de notas {i+1} ({notas_actuales}) es: {promedio:.2f}")
````

## 5. Número indeterminado de argumentos

Cuando la cantidad de argumentos que recibirá una función es variable y no se conoce de antemano, Python ofrece dos operadores especiales: `*args` y `**kwargs`.

### El operador *args (Argumentos Posicionales Variables)

Se utiliza cuando se desea pasar una cantidad variable de argumentos posicionales a una función. Dentro de la función, el parámetro precedido por `*` se comporta como una **tupla** que contiene todos los argumentos posicionales adicionales que se le pasaron.

````python
# Ejemplo sencillo de `*args`, se comporta como una tupla.
def sumar_todos(*numeros):
    print(f"Los argumentos recibidos son: {numeros} y son de tipo: {type(numeros)}")
    return sum(numeros)

print(f"Suma de varios números: {sumar_todos(1, 2, 3, 4)}")
print(f"Suma de un solo número: {sumar_todos()}")
````

````python
def limpiar_varios_textos(*textos):
    return [t.strip().lower() for t in textos]

# Puedes pasarle los argumentos que quieras
resultado = limpiar_varios_textos(" Python ", " DATA ", " Ml ")
print(resultado)
````

### El operador **kwargs (Argumentos de Palabra Clave Variables)

Se utiliza para pasar una cantidad variable de argumentos con nombre (o palabra clave) a una función. Dentro de la función, el parámetro precedido por `**` se comporta como un **diccionario** que contiene todos los argumentos de palabra clave adicionales, donde las claves son los nombres de los argumentos y los valores son sus respectivos valores.

````python
#Ejemplo sencillo de `**kwargs`:
def mostrar_informacion(**detalles):
    print(f"Los argumentos recibidos son: {detalles} y son de tipo: {type(detalles)}")
    for clave, valor in detalles.items():
        print(f"{clave.replace('_', ' ').capitalize()}: {valor}")

print("Información de usuario:")
mostrar_informacion(nombre="Alicia", edad=30, ciudad="San Luis", fecha_de_nacimiento="15 de marzo de 1996" )
````

````python
# kwargs se comporta como un diccionario.
def limpiar_texto_config(texto, **config):
    if config.get("strip", True):
        texto = texto.strip()
    if config.get("lower", True):
        texto = texto.lower()
    if config.get("replace"):
        for viejo, nuevo in config["replace"].items():
            texto = texto.replace(viejo, nuevo)
    return texto

resultado = limpiar_texto_config(
    "  Python Data  ",
    strip=True,
    lower=True,
    replace={"data": "science"}
)

print(resultado)
````

### Orden de Parámetros
1. Obligatorios → 2. *args → 3. Con valor por defecto → 4. **kwargs.

````python
# Ejemplo
def super_funcion(fijo, *args, opcion="predeterminado", **kwargs):
    print(f"Parámetro fijo: {fijo}")
    print(f"Argumentos extra (*args): {args}")
    print(f"Valor por defecto (opcion): {opcion}")
    print(f"Argumentos de diccionario (**kwargs): {kwargs}")

super_funcion("Base", 1, 2, 3, opcion="Cambiado", usuario="admin", log=True)
````

## 6. Argumentos y Mutabilidad
**Diferencia clave:**
* **Inmutables (int, str, tuple):** No cambian fuera de la función.
* **Mutables (list, set, dict):** Si se modifican dentro, cambian fuera.

````python
def modificar_lista(lista, texto):
    lista.append("SCIENCE")   # modificamos el objeto mutable
    texto = texto.upper()     # asignamos un nuevo texto al parametro texto
    print(f"Dentro de la función -> lista: {lista}")
    print(f"Dentro de la función -> texto: {texto}")

mi_lista = ['python', 'data', 'ml']
mi_texto = "ciencia de datos"

modificar_lista(mi_lista, mi_texto)

print(f"\nFuera de la función -> lista: {mi_lista}")
print(f"Fuera de la función -> texto: {mi_texto}")   # los strings son inmutables y texto sigue siendo el mismo
````

### Pasar funciones como parámetros para operaciones en listas

````python
def aplicar_operacion_a_lista(lista, operacion):
    return operacion(lista)

# Probamos la ejecución
numeros = [10, 20, 30, 40, 50]
# funcion definida por nosotros
def multiplicar_por_dos(lista):
    return [x * 2 for x in lista]

lista_duplicada = aplicar_operacion_a_lista(numeros, multiplicar_por_dos)
print(f"Lista duplicada: {lista_duplicada}")
````

````python
# Pasamos la función 'sum' como parámetro
suma_total = aplicar_operacion_a_lista(numeros, sum)
print(f"La suma de los números es: {suma_total}")

# Pasamos la función 'len' como parámetro
cantidad_elementos = aplicar_operacion_a_lista(numeros, len)
print(f"La cantidad de elementos es: {cantidad_elementos}")
````

### Se pueden retornar funciones desde otras funciones (clausura)

````python
#fábrica de funciones
def crear_multiplicador(factor):
    def multiplicar(n):
        return n * factor 
    return multiplicar # devuelve una funcion con el valor de factor guardado internamente
    
#Se crea una instancia de la funcion, asignada a duplicar, donde factor vale 2 (clausura)
duplicar = crear_multiplicador(2)
#Python almacena clausuras en un atributo especial llamado __closure__
print(f"n clausurado en duplicar {duplicar.__closure__[0].cell_contents}")



````

````python
#la función interna busca n (que es 5) y luego busca factor en la clausura de duplicar
print (f" Resultado de duplicar: {duplicar(5)}")
````

````python
#Se crea una instancia de multiplicar, asignada a triplicar, donde n vale 3 (clausura)
triplicar = crear_multiplicador(3)
print(f"n clausurado en triplicar {triplicar.__closure__[0].cell_contents}")
#la función interna busca n (que es 5) y luego busca factor en la clausura de triplicar
print (f" Resultado de triplicar: {triplicar(5)}")
````

````python
# ¿Qué pasará aquí?
nueva_funcion = crear_multiplicador(10)
print(f"n clausurado en nueva_funcion {nueva_funcion.__closure__[0].cell_contents}")
print (f" Resultado de nueva funcion: {nueva_funcion(duplicar(2))}")
````

### Otro ejemplo de Clausura: Generador de Contadores

````python
def crear_contador(inicio=0):
    """
    Crea y retorna una función de contador.
    La función interna 'incrementar' recuerda el valor de 'inicio'.
    """
    count = inicio
    def incrementar():
        nonlocal count # Declara que 'count' no es una variable local, sino de la función externa
        count += 1
        return count
    return incrementar


# Creamos dos contadores diferentes
contador1 = crear_contador(inicio=5)
contador2 = crear_contador(inicio=100)

print("Contador 1:")
print(f"Primero: {contador1()}") # 6
print(f"Segundo: {contador1()}") # 7

print("\nContador 2:")
print(f"Primero: {contador2()}") # 101
print(f"Segundo: {contador2()}") # 102
print(f"Tercero: {contador2()}") # 103
````

## Introducción a Docstrings y Typehints

Los `docstrings` (cadenas de documentación) son literales de cadena que aparecen como la primera instrucción en una definición de módulo, función, clase o método. Sirven para explicar el propósito del código, describir sus argumentos, qué hace y qué valor retorna. Son cruciales para:
- **Documentación interna:** Ayudan a los desarrolladores a entender rápidamente cómo usar una función o clase sin tener que leer todo el código fuente.
- **Herramientas de ayuda:** IDEs y herramientas como `help()` de Python los utilizan para mostrar información contextual sobre las funciones.
- **Generación automática de documentación:** Facilitan la creación de documentación de proyectos de forma automatizada.

Los `typehints` (sugerencias de tipo) son una característica de Python 3.5+ que permite a los desarrolladores indicar los tipos de datos esperados para los parámetros de una función y su valor de retorno. No obligan al intérprete de Python a verificar los tipos en tiempo de ejecución (Python sigue siendo dinámicamente tipado), pero son extremadamente útiles para:
- **Legibilidad y comprensión:** Hacen el código más claro sobre qué tipo de datos espera una función y qué devolverá.
- **Detección temprana de errores:** Herramientas de análisis estático de código (como `mypy`) pueden usarlos para encontrar errores de tipo antes de ejecutar el código, lo que facilita la depuración.
- **Mejora del autocompletado:** Los IDEs pueden proporcionar sugerencias de código más precisas y útiles.
