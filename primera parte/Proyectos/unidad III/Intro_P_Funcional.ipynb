# Programación Funcional en Python

La **programación funcional** es un paradigma de programación que trata la computación como la evaluación de funciones matemáticas y evita el cambio de estado y los datos mutables. Se enfoca en el uso de funciones puras, lo que significa que una función siempre producirá el mismo resultado para las mismas entradas y no tendrá efectos secundarios.

Este enfoque promueve un código más conciso, modular, predecible y fácil de probar. En Python, aunque no es un lenguaje puramente funcional, podemos aplicar muchos de sus principios y herramientas para escribir código más limpio y eficiente.

## Objetivos

## Funciones como objetos

En Python, las funciones son **ciudadanos de primera clase**, lo que significa que pueden ser tratadas como cualquier otra variable u objeto. Esto implica que una función puede:

*   **Asignarse a una variable:** Puedes guardar una función en una variable y luego usar esa variable para llamarla.
    ```python
    def saludo():
        print("Hola")

    mi_funcion = saludo
    mi_funcion()
    ```
*   **Pasarse como argumento a otra función:** Las funciones que aceptan otras funciones como argumentos se conocen como **funciones de orden superior**.
    ```python
    def aplicar_operacion(funcion, a, b):
        return funcion(a, b)

    def sumar(x, y):
        return x + y

    resultado = aplicar_operacion(sumar, 5, 3)
    print(resultado) # Salida: 8
    ```
*   **Retornarse como resultado de otra función:** Una función puede generar y devolver otra función.
    ```python
    def crear_multiplicador(n):
        def multiplicador(x):
            return x * n
        return multiplicador

    duplicar = crear_multiplicador(2)
    print(duplicar(5)) # Salida: 10
    ```

Esta característica es fundamental para la **programación funcional**, ya que permite mayor flexibilidad y modularidad en el diseño del código, facilitando la creación de abstracciones poderosas y la composición de funciones.

````python
def cuadrado(x):
    return x*x

f = cuadrado
print(f(5))
````

## Lambda (funciones anónimas)

Las funciones `lambda` en Python son **funciones pequeñas y anónimas** (es decir, no tienen un nombre formal) que pueden tomar cualquier número de argumentos, pero solo pueden tener una única expresión. Se definen utilizando la palabra clave `lambda`.

Su sintaxis general es:
`lambda argumentos: expresión`

**Características clave:**
*   **Anónimas:** No necesitan un nombre definido con `def`.
*   **Una sola expresión:** La expresión se evalúa y se devuelve automáticamente. No se pueden usar sentencias complejas como `if`, `for` o `return` explícitamente dentro de una lambda.
*   **Concisión:** Son ideales para tareas sencillas que requieren una función de corta duración, como ordenar listas, o como funciones de devolución de llamada.

**Uso común:**
Son particularmente útiles cuando se usan junto con funciones de orden superior como `map()`, `filter()`, `sorted()`, o `reduce()`, donde se necesita una función rápida para una operación específica sin la necesidad de definirla formalmente en otro lugar del código.

````python
cuadrado = lambda x: x*x
print(cuadrado(6))
````

````python
resultado=lambda a,b,c: a + b + c
print(resultado(2,4,6))
````

**Clausura - Lambda**

Una función lambda puede definirse dentro de otra función y ser retornada.
La función resultante mantiene acceso a las variables del entorno donde fue creada.

````python
def crear_multiplicador(factor):
    return lambda a : a * factor

doble= crear_multiplicador(2)
print (doble(10))
````

## map(): transformar datos

La función incorporada `map()` de Python aplica una función dada a **cada elemento** de una secuencia (como una lista, tupla o cualquier iterable) y devuelve un **iterador** con los resultados de esas aplicaciones. Es una herramienta poderosa para transformar colecciones de datos de forma concisa y funcional.

Su sintaxis es:
`map(funcion, iterable)` --> map object

**Parámetros:**
*   `funcion`: La función que se aplicará a cada elemento del iterable. Puede ser una función definida con `def`, una función `lambda`, o una función incorporada.
*   `iterable`: Uno o más objetos iterables (como listas, tuplas, cadenas) cuyos elementos serán procesados por la función.

**Funcionamiento:**
`map()` toma un elemento del iterable, le aplica la `funcion`, y el resultado se convierte en el siguiente elemento del iterador devuelto. Este proceso continúa hasta que se procesan todos los elementos del iterable.

Para obtener una lista con los resultados, es común convertir el iterador `map` a una lista utilizando `list()`:
`resultado = list(map(funcion, iterable))`

**Ventajas:**
*   **Código conciso:** Permite transformar colecciones de datos en una sola línea.
*   **Eficiencia:** Es a menudo más eficiente que un bucle `for` explícito para tareas de transformación simples, especialmente con grandes conjuntos de datos, ya que trabaja con iteradores y no crea listas intermedias completas en memoria de inmediato.

````python
datos = [1,2,3,4]
resultado=map(lambda x: x*2, datos)

print(type(resultado))

#Si no lo guardás en una lista… lo perdés después de usarlo
print(list(resultado))
````

## filter(): filtrar datos

La función incorporada `filter()` de Python construye un **iterador** a partir de los elementos de una secuencia (iterable) para los cuales una función dada devuelve `True`. Es ideal para seleccionar o filtrar elementos de una colección que cumplen con una condición específica.

Su sintaxis es:
`filter(funcion, iterable)` --> filter object

**Parámetros:**
*   `funcion`: Una función que prueba cada elemento del iterable. Esta función debe devolver un valor booleano (`True` o `False`). Si se pasa `None` como función, `filter()` eliminará los elementos que sean `False` en un contexto booleano (como `0`, `None`, `False`, `''`, `[]`).
*   `iterable`: Uno o más objetos iterables (como listas, tuplas, cadenas) cuyos elementos serán probados por la función.

**Funcionamiento:**
`filter()` aplica la `funcion` a cada elemento del `iterable`. Si la `funcion` devuelve `True` para un elemento, ese elemento se incluye en el iterador resultante. Si devuelve `False`, el elemento se descarta.

Al igual que con `map()`, para obtener una lista con los elementos filtrados, es común convertir el iterador `filter` a una lista:
`resultado = list(filter(funcion, iterable))`

**Ventajas:**
*   **Claridad:** Proporciona una manera declarativa y fácil de leer para filtrar datos.
*   **Eficiencia:** Al igual que `map()`, trabaja con iteradores, lo que puede ser más eficiente en memoria para grandes conjuntos de datos que construir una nueva lista con un bucle `for` y condicionales.

````python
datos = [10,25,30,45]

resultado = filter(lambda x: x>20, datos)
print(type(resultado))

print(list(resultado))
print(list(resultado))
````
