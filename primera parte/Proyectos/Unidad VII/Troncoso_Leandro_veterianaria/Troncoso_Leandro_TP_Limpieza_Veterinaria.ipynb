# Práctico Limpieza de Datos con Pandas - Clínica Veterinaria

**Contexto:** Una clínica veterinaria registró información de mascotas atendidas durante distintas semanas. El dataset contiene inconsistencias, valores faltantes y datos incorrectos.


```python
import pandas as pd
import numpy as np
```

## Creación del Dataset

```python
mascotas = pd.DataFrame({
    "ID_Consulta": [201, 202, 203, 204, 204, 206, 207, 208, 209, 210],
    "NombreMascota": [
        "Luna", "Max", "Milo", "Nina", "Nina",
        None, "Rocky", "Simba", "Mora", "Luna"
    ],
    "Especie": [
        "Perro", "gato", "GATO", "Perro",
        "Perro", "Conejo", "conejo", "Perro", None, "PERRO"
    ],
    "PesoKg": [
        "12.5", "4.2", "error", "8.1", "8.1",
        "2.4", None, "15", "-3", "13"
    ],
    "Edad": [
        5, 3, np.nan, 2, 2, 1, 4, -1, 6, 5
    ],
    "FechaConsulta": [
        "01/04/2025", "2025-04-02", "03-04-2025", "04/04/2025",
        "04/04/2025", "05/04/2025", "06/04/2025", "07/04/2025",
        "08/04/2025", "09/04/2025"
    ]
})

mascotas
```

## Actividad 4: Identificar valores faltantes, duplicados y tipos incorrectos

```python
# Conteo y porcentaje de nulos por columna
nulos = mascotas.isnull().sum()
porcentaje = (nulos / len(mascotas) * 100).round(1)

resumen_nulos = pd.DataFrame({
    "Nulos": nulos,
    "Porcentaje (%)": porcentaje
})

print("=== VALORES FALTANTES ===")
print(resumen_nulos)
print()
```

```python
# Duplicados
print(f"=== DUPLICADOS ===")
print(f"Filas duplicadas exactas: {mascotas.duplicated().sum()}")
print()
print("Filas duplicadas:")
print(mascotas[mascotas.duplicated(keep=False)])
```

```python
# Tipos de datos
print("=== TIPOS DE DATOS ===")
print(mascotas.dtypes)
print()
print("Observaciones:")
print("- PesoKg es 'object' (texto) en lugar de numérico")
print("- FechaConsulta es 'object' en lugar de datetime")
```

```python
# Frecuencia de valores en Especie para ver inconsistencias
print("Especies:")
print(mascotas["Especie"].value_counts(dropna=False))
```

---
## Actividad 7: Unificar formatos de texto en la columna Especie

Siguiendo el paso 2 del cuaderno de clase: normalizar cadenas de texto con `str.strip()` y `str.capitalize()`.

```python
# Trabajamos sobre una copia
df_limpio = mascotas.copy()

# Normalizar Especie: eliminar espacios y capitalizar
df_limpio["Especie"] = df_limpio["Especie"].str.strip().str.capitalize()

print("Especies únicas después de limpiar:")
print(df_limpio["Especie"].value_counts(dropna=False))
```

---
## Actividad 5: Detectar y reemplazar valores nulos

Siguiendo la estrategia de clase:
- Variables categóricas → imputar con **moda**
- Variables numéricas → imputar con **media** o **mediana**

```python
# Nulos actuales después de la limpieza de texto
print("Nulos actuales:")
print(df_limpio.isnull().sum())
print()
```

```python
# NombreMascota (categórica) → imputar con moda
moda_nombre = df_limpio["NombreMascota"].mode()[0]
df_limpio["NombreMascota"] = df_limpio["NombreMascota"].fillna(moda_nombre)
print(f"NombreMascota imputada con moda: '{moda_nombre}'")
```

```python
# Especie (categórica) → imputar con moda
moda_especie = df_limpio["Especie"].mode()[0]
df_limpio["Especie"] = df_limpio["Especie"].fillna(moda_especie)
print(f"Especie imputada con moda: '{moda_especie}'")
```

```python
# Edad (numérica) → imputar con mediana (más robusta ante outliers, como se vio en clase)
mediana_edad = df_limpio["Edad"].median()
df_limpio["Edad"] = df_limpio["Edad"].fillna(mediana_edad)
print(f"Edad imputada con mediana: {mediana_edad}")
```

```python
# Verificamos que no queden nulos (PesoKg se trata en la actividad 8)
print("Nulos restantes:")
print(df_limpio.isnull().sum())
```

---
## Actividad 6: Eliminar registros duplicados

Usando `drop_duplicates(keep='first')` como se vio en clase.

```python
print(f"Filas antes de eliminar duplicados: {len(df_limpio)}")

# Eliminar duplicados exactos en todas las columnas
df_limpio = df_limpio.drop_duplicates(keep="first")
print(f"Filas después de eliminar duplicados exactos: {len(df_limpio)}")

# Resetear el índice después de eliminar filas
df_limpio.reset_index(drop=True, inplace=True)
print("Índice del DataFrame reseteado.")
```

---
## Actividad 8: Convertir PesoKg a tipo numérico

Usando `pd.to_numeric(errors='coerce')` como se vio en clase. Los valores no convertibles (como 'error') se transforman en NaN.

```python
print("Antes - tipo de dato PesoKg:", df_limpio["PesoKg"].dtype)
print("Valores actuales:")
print(df_limpio["PesoKg"].values)
print()
```

```python
# Convertir a numérico (errors='coerce' convierte 'error' y similares a NaN)
df_limpio["PesoKg"] = pd.to_numeric(df_limpio["PesoKg"], errors="coerce")

print("Después - tipo de dato PesoKg:", df_limpio["PesoKg"].dtype)
print(df_limpio["PesoKg"])
```

```python
# Imputar los NaN generados en PesoKg con la mediana
mediana_peso = df_limpio["PesoKg"].median()
df_limpio["PesoKg"] = df_limpio["PesoKg"].fillna(mediana_peso)
print(f"PesoKg: NaN imputados con mediana: {mediana_peso}")
```

---
## Actividad 9: Convertir FechaConsulta a datetime

Usando `pd.to_datetime(dayfirst=True, format='mixed')` como se vio en clase para manejar distintos formatos de fecha.

```python
print("Antes - tipo de dato FechaConsulta:", df_limpio["FechaConsulta"].dtype)
print("Valores únicos:", df_limpio["FechaConsulta"].unique())
```

```python
# Convertir a datetime (dayfirst=True y format='mixed' para manejar distintos formatos)
df_limpio["FechaConsulta"] = pd.to_datetime(df_limpio["FechaConsulta"], dayfirst=True, format='mixed')

print("Después - tipo de dato:", df_limpio["FechaConsulta"].dtype)
print(df_limpio["FechaConsulta"])
```

---
## Actividad 10: Detectar y corregir valores inválidos

Siguiendo el paso 5 de clase (Identificación de Valores Sospechosos): usamos `describe()` para inspeccionar rangos y detectar valores fuera de lugar. Luego usamos máscaras booleanas para reemplazar por la mediana.

```python
# Inspeccionar rangos con describe()
print("=== Estadísticas para detectar valores sospechosos ===")
print(df_limpio[["PesoKg", "Edad"]].describe())
```

```python
# PesoKg: valores negativos son inválidos
print("Valores de PesoKg negativos:")
print(df_limpio[df_limpio["PesoKg"] < 0])
print()

# Reemplazar valores negativos de PesoKg por la mediana (máscara booleana)
mediana_peso = df_limpio.loc[df_limpio["PesoKg"] > 0, "PesoKg"].median()
df_limpio.loc[df_limpio["PesoKg"] < 0, "PesoKg"] = mediana_peso
print(f"PesoKg negativos reemplazados por mediana: {mediana_peso}")
```

```python
# Edad: valores negativos son inválidos
print("Valores de Edad negativos:")
print(df_limpio[df_limpio["Edad"] < 0])
print()

# Reemplazar valores negativos de Edad por la mediana (máscara booleana)
mediana_edad = df_limpio.loc[df_limpio["Edad"] > 0, "Edad"].median()
df_limpio.loc[df_limpio["Edad"] < 0, "Edad"] = mediana_edad
print(f"Edad negativos reemplazados por mediana: {mediana_edad}")
```

---
## Validación Final

Como se vio en clase (Paso 7), hacemos una revisión final para confirmar que el dataset quedó limpio.

```python
print("=== RESULTADOS ===")
print(f"Filas originales : {len(mascotas)}")
print(f"Filas limpias    : {len(df_limpio)}")
print(f"\nNulos restantes  : {df_limpio.isnull().sum().sum()}")
print(f"Duplicados rest. : {df_limpio.duplicated().sum()}")
print(f"\nTipos de datos:")
print(df_limpio.dtypes)
```

```python
# Aserciones (vistas en clase)
assert not df_limpio.isnull().any().any(), "Se encontraron valores nulos en el DataFrame"
assert not df_limpio.duplicated().any(), "Existen filas duplicadas exactas en el dataset"

print("Todas las aserciones pasaron correctamente.")
```
