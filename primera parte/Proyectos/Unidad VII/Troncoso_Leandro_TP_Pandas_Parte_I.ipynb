## PARTE 4 — Operaciones y nuevas columnas

Vamos a enriquecer el DataFrame creando columnas nuevas a partir de las existentes.

**Ejercicio 4.1** — Creá una nueva columna llamada `bmi_category` que clasifique a cada cliente según su BMI:

- `"Bajo peso"` si `bmi < 18.5`
- `"Normal"` si `18.5 <= bmi <= 24.9`  
- `"Sobrepeso"` si `25 <= bmi <= 29.9`
- `"Obesidad"` si `bmi >= 30`

*Pista: podés usar `.apply()` o  `pd.cut()` o Resolver de ambas formas y justificar cual es mejor.*

```python
# Tu código aquí

# Con apply (mapea directamente las condiciones del enunciado)
def clasificar_bmi(bmi):
    if bmi < 18.5:
        return 'Bajo peso'
    elif bmi <= 24.9:
        return 'Normal'
    elif bmi <= 29.9:
        return 'Sobrepeso'
    else:
        return 'Obesidad'

df['bmi_category'] = df['bmi'].apply(clasificar_bmi)

# Con pd.cut (alternativa más compacta)
# df['bmi_category'] = pd.cut(df['bmi'],
#     bins=[0, 18.5, 25, 30, float('inf')],
#     labels=['Bajo peso', 'Normal', 'Sobrepeso', 'Obesidad'],
#     right=False)
```

**Ejercicio 4.2**  — Mostrá cuántos clientes hay en cada categoría bmi_category

```python
# Tu código aquí

df['bmi_category'].value_counts()
```

**Ejercicio 4.3** — Creá una nueva columna llamada `charges_per_child` que represente cuánto paga el cliente por cada hijo en el seguro.  
Para los clientes **sin hijos**, el valor debe ser `NaN` (recordá: dividir entre 0 no tiene sentido).

*Pista: Aunque podrías resolver este ejercicio usando .apply(lambda...), en la práctica profesional se prefiere np.where().*

```python
# Tu código aquí

df['charges_per_child'] = np.where(df['children'] > 0, df['charges'] / df['children'], np.nan)
df[['children', 'charges', 'charges_per_child']].head(10)
```

## PARTE 5 — Filtrado de datos

Aplicá condiciones para seleccionar subconjuntos del DataFrame.

**Ejercicio 5.1** — Filtrá los clientes que son **fumadores**. ¿Cuántos hay?

```python
# Tu código aquí

fumadores = df[df['smoker'] == 'yes']
print(f"Hay {len(fumadores)} clientes fumadores")
```

**Ejercicio 5.2** — Filtrá los clientes que son **fumadores Y tienen obesidad** (BMI >= 30). Mostrá solo las columnas `age`, `sex`, `bmi`, `smoker` y `charges`.

```python
# Tu código aquí

filtro = (df['smoker'] == 'yes') & (df['bmi'] >= 30)
df[filtro][['age', 'sex', 'bmi', 'smoker', 'charges']]
```

**Ejercicio 5.3** — Filtrá los clientes de la región `'southeast'` que tienen **más de 2 hijos** y cuyos gastos médicos superan los **$15,000**.

```python
# Tu código aquí

filtro = (df['region'] == 'southeast') & (df['children'] > 2) & (df['charges'] > 15000)
df[filtro]
```

**Ejercicio 5.4** — ¿Cuál es el gasto médico promedio de los clientes **no fumadores**? ¿Y el de los **fumadores**?  
Imprimí ambos valores con un mensaje descriptivo.

```python
# Tu código aquí

prom_no_fumadores = df[df['smoker'] == 'no']['charges'].mean()
prom_fumadores = df[df['smoker'] == 'yes']['charges'].mean()
print(f"Gasto promedio no fumadores: ${prom_no_fumadores:.2f}")
print(f"Gasto promedio fumadores:    ${prom_fumadores:.2f}")
```
