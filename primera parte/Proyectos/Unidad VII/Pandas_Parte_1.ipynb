- **Facilita el Manejo de Datos**: Pandas simplifica el manejo de datos, para enfocarnos en el análisis.

## 1. Introducción a Pandas: Series y DataFrames

Pandas tiene dos estructuras de datos principales:
- **`Series`**: arreglo unidimensional etiquetado (como una columna)
- **`DataFrame`**: tabla bidimensional con etiquetas en filas y columnas

````python
datos = {"nombre": ["Pedro", "Juan", "Lorena"], "edad": [25, 39, 33]}
datos
````

````python
# ----- DATAFRAME -----
# Desde diccionario
df = pd.DataFrame(datos)
df
````

````python
type(df)
````

### Series
El segundo tipo de datos que incorpora Pandas, son las series. Las series son las columnas de los DataFrames.

En el caso de df tenemos 2 series (la serie nombre y la serie edad), ya que la primera columna, aunque parece una serie más, solo se trata de los índices de cada registro o fila de nuestro dataframe.

Podemos acceder a las series de nuestro objeto dataframe, usando el sistema de corchetes, o el de notación de punto.

````python
df["nombre"]
````

````python
# ----- SERIES -----
# Desde lista
temperaturas = pd.Series([22.5, 25.1, 19.8, 28.3, 23.0],
                          index=['lun', 'mar', 'mié', 'jue', 'vie'],
                          name='Temperatura (°C)')
print('Serie de temperaturas:')
print(temperaturas)
print(f'\nTipo: {type(temperaturas)}')
print(f'dtype: {temperaturas.dtype}')
````

````python
# ----- DATAFRAME -----
# Desde diccionario
np.random.seed(42)

data = {
    'nombre': ['Ana', 'Carlos', 'María', 'Luis', 'Elena', 'Pedro', 'Sofía'],
    'edad': [28, 34, 29, 45, 31, 38, 26],
    'ciudad': ['Salta', 'Mendoza', 'Buenos Aires', 'Córdoba', 'Buenos Aires', 'Buenos Aires', 'Santa Fe'],
    'salario': [45000, 62000, 48000, 71000, 55000, 59000, 43000],
    'experiencia_años': [3, 8, 4, 18, 6, 12, 2],
    'satisfaccion': [4, 3, 5, 4, 2, 3, 5]  # 1-5
}

df = pd.DataFrame(data)
print('DataFrame de empleados:')
df
````

````python
# Atributos esenciales de un DataFrame
print('--- Atributos del DataFrame ---')
print(f'Forma (filas, columnas): {df.shape}')
print(f'Columnas: {df.columns.tolist()}')
print(f'Índice: {df.index.tolist()}')
print(f'Tipos de datos:\n{df.dtypes}')
````

# `pd.read_csv()` — Parámetros de configuración

---

## Archivo

| Parámetro | Tipo | Default | Descripción |
|-----------|------|---------|-------------|
| `filepath_or_buffer` | `str / Path / IO` | — | Ruta al archivo, URL o cualquier objeto con método `read()`. Único argumento obligatorio. |
| `sep` | `str` | `','` | Separador de campos. Puede ser carácter, string o regex. `None` = el motor lo infiere. |
| `delimiter` | `str` | `None` | Alias de `sep`. |
| `encoding` | `str` | `None` | Codificación del archivo: `'utf-8'`, `'latin-1'`, `'cp1252'`, etc. |
| `compression` | `str / dict` | `'infer'` | Compresión: `'infer'`,`'gzip'`,`'bz2'`,`'zip'`,`'xz'`,`'zstd'`,`'tar'`. |
| `storage_options` | `dict` | `None` | Opciones para sistemas remotos (S3, GCS, etc.). |

---

## Encabezado

| Parámetro | Tipo | Default | Descripción |
|-----------|------|---------|-------------|
| `header` | `int / list / None` | `'infer'` | Fila(s) usadas como nombres de columna. `0`=primera fila, `None`=sin encabezado. |
| `names` | `list` | `None` | Lista de nombres de columna a asignar. Combinar con `header=0` para reemplazar el existente. |
| `prefix` | `str` | `None` | Prefijo para columnas sin nombre. Ej: `prefix='col'` genera `col0, col1...` *(deprecado)* |

---

## Índice

| Parámetro | Tipo | Default | Descripción |
|-----------|------|---------|-------------|
| `index_col` | `int / str / list / False` | `None` | Columna(s) a usar como índice. `False` evita que la primera columna sea índice automáticamente. |

---

## Columnas

| Parámetro | Tipo | Default | Descripción |
|-----------|------|---------|-------------|
| `usecols` | `list / callable` | `None` | Selecciona un subconjunto de columnas. Acepta lista de nombres, índices o función lambda. |
| `dtype` | `type / dict` | `None` | Tipo de dato por columna. Ej: `{'edad':int, 'precio':float}`. Acelera lectura y ahorra memoria. |
| `converters` | `dict` | `None` | Funciones personalizadas al leer. Ej: `{'fecha': pd.to_datetime}`. |

---

## Filas

| Parámetro | Tipo | Default | Descripción |
|-----------|------|---------|-------------|
| `skiprows` | `int / list / callable` | `None` | Filas a omitir. `skiprows=2` salta las 2 primeras; `[0,3]` salta filas 0 y 3. |
| `skipfooter` | `int` | `0` | Filas al final a ignorar. Solo con `engine='python'`. |
| `nrows` | `int` | `None` | Número máximo de filas a leer. Ideal para previsualizar archivos grandes. |
| `skip_blank_lines` | `bool` | `True` | Si `True`, ignora líneas en blanco en lugar de interpretarlas como `NaN`. |

---

## Valores faltantes

| Parámetro | Tipo | Default | Descripción |
|-----------|------|---------|-------------|
| `na_values` | `scalar / list / dict` | `None` | Valores adicionales a reconocer como `NaN` (además de `''`, `'NA'`, `'NaN'`, `'null'`, etc.). |
| `keep_default_na` | `bool` | `True` | Si `False`, solo usa `na_values` y no los predefinidos de pandas. |
| `na_filter` | `bool` | `True` | Si `False`, desactiva detección de `NaN`. Mejora rendimiento si no hay faltantes. |

---

## Tipos y fechas

| Parámetro | Tipo | Default | Descripción |
|-----------|------|---------|-------------|
| `parse_dates` | `bool / list / dict` | `False` | Parsea fechas. `True`=todas; lista de columnas; dict para combinar columnas. |
| `dayfirst` | `bool` | `False` | Si `True`, interpreta fechas ambiguas como `DD/MM/YYYY`. Importante para datos europeos. |
| `keep_date_col` | `bool` | `False` | Si `True`, conserva columnas originales usadas para construir una fecha compuesta. |
| `date_parser` | `callable` | `None` | Función personalizada para parsear fechas. *(deprecado)* |
| `infer_datetime_format` | `bool` | `False` | Infiere formato de fecha para acelerar el parsing. *(deprecado en pandas 2.x)* |

---

## Rendimiento

| Parámetro | Tipo | Default | Descripción |
|-----------|------|---------|-------------|
| `chunksize` | `int` | `None` | Lee en fragmentos de N filas. Devuelve un `TextFileReader` iterable. |
| `iterator` | `bool` | `False` | Si `True`, devuelve `TextFileReader` para llamar `.get_chunk(n)` manualmente. |
| `low_memory` | `bool` | `True` | Procesa en trozos para ahorrar memoria, pero puede inferir tipos mixtos incorrectamente. |
| `memory_map` | `bool` | `False` | Si `True`, usa `mmap` del sistema. Más rápido en archivos locales grandes. |

---

## Motor

| Parámetro | Tipo | Default | Descripción |
|-----------|------|---------|-------------|
| `engine` | `str` | `None` | Motor: `'c'` (rápido), `'python'` (más features), `'pyarrow'` (experimental). |
| `on_bad_lines` | `str / callable` | `'error'` | Líneas malformadas: `'error'` lanza excepción, `'warn'` avisa, `'skip'` omite. |

---

## Texto

| Parámetro | Tipo | Default | Descripción |
|-----------|------|---------|-------------|
| `quotechar` | `str` | `'"'` | Carácter para delimitar campos que contienen el separador. |
| `quoting` | `int` | `0` | `0`=MINIMAL, `1`=ALL, `2`=NONNUMERIC, `3`=NONE. |
| `doublequote` | `bool` | `True` | Si `True`, `""` dentro de un campo se interpreta como una sola comilla literal. |
| `escapechar` | `str` | `None` | Carácter de escape para el `quotechar` o el separador. |
| `comment` | `str` | `None` | Carácter que indica inicio de comentario. Esas líneas se ignoran. |
| `thousands` | `str` | `None` | Separador de miles. Ej: `thousands=','` para leer `'1,000,000'` como `1000000`. |
| `decimal` | `str` | `'.'` | Carácter decimal. Usar `decimal=','` para archivos europeos. |
| `delim_whitespace` | `bool` | `False` | Si `True`, usa whitespace como delimitador. *(deprecado, usar `sep='\s+'`)* |

---

## Booleanos

| Parámetro | Tipo | Default | Descripción |
|-----------|------|---------|-------------|
| `true_values` | `list` | `None` | Valores adicionales a interpretar como `True`. |
| `false_values` | `list` | `None` | Valores adicionales a interpretar como `False`. |

---

## Avanzado

## 3. Exploración Inicial de Datos (EDA Básico)

Los primeros pasos al recibir cualquier dataset.

#### Explicación Breve de Métodos de Exploración

*   **`.head(n)`:** Muestra las primeras `n` filas del DataFrame. Por defecto, `n` es 5. Es útil para una vista rápida de la estructura y el contenido inicial de los datos.

*   **`.info()`:** Proporciona un resumen conciso del DataFrame, incluyendo el número de entradas, el número de columnas, los tipos de datos de cada columna, el número de valores no nulos y el uso de memoria. Es crucial para identificar valores faltantes y tipos de datos incorrectos.

*   **`.describe(include='all')`:** Genera estadísticas descriptivas de las columnas. Para columnas numéricas, incluye conteo, media, desviación estándar, valores mínimos y máximos, y cuartiles. Con `include='all'`, también proporciona estadísticas para columnas categóricas como el conteo, el número de valores únicos, el valor más frecuente (`top`) y su frecuencia (`freq`).

*   **`.value_counts()`:** Es un método de Series que devuelve una Serie que contiene el conteo de valores únicos en la columna, en orden descendente por defecto. Es muy útil para entender la distribución de valores en una columna categórica o discreta.

````python
# Creamos un dataset más grande y realista para EDA
np.random.seed(42)
n = 200

ciudades = ['Buenos Aires', 'Córdoba', 'Mendoza', 'Santa Fe', 'Salta']
departamentos = ['Ingeniería', 'Marketing', 'Ventas', 'RRHH', 'Finanzas']

df_emp = pd.DataFrame({
    'id_empleado': range(1, n+1),
    'nombre': [f'Empleado_{i}' for i in range(1, n+1)],
    'edad': np.random.randint(22, 60, n),
    'ciudad': np.random.choice(ciudades, n),
    'departamento': np.random.choice(departamentos, n),
    'salario': np.random.normal(55000, 12000, n).round(0),
    'experiencia_años': np.random.randint(0, 30, n),
    'satisfaccion': np.random.randint(1, 6, n),
    'activo': np.random.choice([True, False], n, p=[0.85, 0.15])
})

# Introducir algunos valores nulos (simulando datos reales)
df_emp.loc[df_emp.sample(20).index, 'salario'] = np.nan
df_emp.loc[df_emp.sample(10).index, 'satisfaccion'] = np.nan

print(f'Dataset generado: {df_emp.shape}')
````

````python
# 1. Primeras y últimas filas
print('=== PRIMERAS 5 FILAS ===')
display(df_emp.head())

print('\n=== ÚLTIMAS 3 FILAS ===')
display(df_emp.tail(3))

print('\n=== MUESTRA ALEATORIA ===')
display(df_emp.sample(3, random_state=42))
````

````python
print('=== CONTEO DE VALORES ÚNICOS POR CIUDAD ===')
df_emp['ciudad'].value_counts()
````

````python
# 2. Información general del DataFrame
print('=== INFO ===')
df_emp.info()
````

````python
# 3. Estadísticas descriptivas
print('=== ESTADÍSTICAS DESCRIPTIVAS ===')
df_emp.describe(include='all').round(2)
````

## 4. Indexación, Selección y Filtrado

Formas de seleccionar subconjuntos del DataFrame: `[]`, `.loc[]`, `.iloc[]`, y filtros booleanos.

#### Explicación Breve de Métodos de Selección

*   **`.loc[]` (Selección basada en etiquetas):** Se utiliza para seleccionar filas y columnas por sus **etiquetas** (nombres de índice y nombres de columna). Permite un filtrado más expresivo y legible cuando las etiquetas son significativas.

*   **`.iloc[]` (Selección basada en posición entera):** Se utiliza para seleccionar filas y columnas por sus **posiciones enteras** (índices numéricos de 0 a N-1). Es útil cuando necesitas seleccionar datos basándote en su ubicación en el DataFrame.

*   **`.query()` (Selección basada en expresiones de cadena):** Permite filtrar filas de un DataFrame utilizando una **expresión booleana en forma de cadena de texto**. A menudo es más legible que los filtros booleanos tradicionales con múltiples condiciones.

*   **`.isin()` (Filtrado por pertenencia a una lista):** Es un método de Series que devuelve un valor booleano indicando si cada elemento de la Serie **está contenido en los valores de una lista o Series pasadas como argumento**. Es muy útil para filtrar filas donde una columna toma uno de varios valores posibles.

````python
# ----- Selección de columnas -----
# Una columna → Series
salarios = df_emp['salario']
print(salarios)
print(type(salarios))

````

````python
# Varias columnas → DataFrame
subset = df_emp[['nombre', 'departamento', 'salario']]
subset.head(3)
````

````python
# ----- .loc[] → por ETIQUETAS -----
# Filas 0-4, columnas específicas
df_emp.loc[0:4, ['nombre', 'ciudad', 'salario']]
````

````python
# ----- .iloc[] → por POSICIÓN numérica -----
# Filas 0-4, columnas 0,2,5
df_emp.iloc[0:5, [0, 2, 5]]
````

````python
# ----- Filtros booleanos -----
# Un filtro
ingenieros = df_emp[df_emp['departamento'] == 'Ingeniería']
print(f'Ingenieros: {len(ingenieros)}')

# Múltiples condiciones (& = AND, | = OR, ~ = NOT)
filtro = (df_emp['salario'] > 60000) & (df_emp['activo'] == True)
bien_pagados_activos = df_emp[filtro]
print(f'Activos con salario > 60k: {len(bien_pagados_activos)}')

# .query() → sintaxis más legible
resultado = df_emp.query('salario > 60000 and edad < 35')
print(f'Salario > 60k y edad < 35: {len(resultado)}')
````

````python
# ----- .isin() y .str -----
# isin: filtrar por lista de valores
norte = df_emp[df_emp['ciudad'].isin(['Buenos Aires', 'Santa Fe', 'Salta'])]
print(f'Empleados en Buenos Aires/Santa Fe/Salta: {len(norte)}')

# str accessor: operaciones de string en Series de texto
bsas_emp = df_emp[df_emp['ciudad'].str.upper() == 'BUENOS AIRES']
print(f'Empleados en Buenos Aires: {len(bsas_emp)}')
print(f'Ciudades en mayúsculas: {df_emp["ciudad"].str.upper().unique()}')
````
