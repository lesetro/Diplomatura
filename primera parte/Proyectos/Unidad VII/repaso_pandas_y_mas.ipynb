### `pd.cut`

`pd.cut` es una función fundamental en Pandas que se utiliza para **segmentar y ordenar valores de datos en discretos 'bins' (intervalos)**. Es ideal para transformar datos numéricos continuos en categorías ordinales o agrupar datos en rangos específicos, siendo más eficiente y legible que `apply(lambda x: ...) ` para esta tarea.

**Sintaxis:**
`pd.cut(x, bins, right=True, labels=None, include_lowest=False)`

**Parámetros clave:**

* x: serie

* Bins: Define los límites.

* Labels: Los nombres de las categorías.

* Right=False: Maneja si el límite superior se incluye o no.


```python
# El primer rango es de 0 a 200 (bajo), 200 a 240 (normal), 240 a infinito (alto)
bins = [0, 200, 240, float('inf')]
labels = ['bajo', 'normal', 'alto']
# right=False hace que el intervalo sea [200, 240), es decir, incluye el 200 pero no el 240
df['categoria_colesterol'] = pd.cut(df['colesterol'], bins=bins, labels=labels, right=False)
```

### Explicación de `df.drop()`

El método `df.drop()` se utiliza para **eliminar filas o columnas** de un DataFrame. Es importante recordar que, por defecto, `drop()` devuelve una *copia* del DataFrame con los elementos eliminados, no modifica el DataFrame original a menos que se use el parámetro `inplace=True`.

**Parámetros clave:**

*   **`labels`**: Corresponde al nombre(s) de la fila(s) o columna(s) que deseas eliminar. Puede ser una etiqueta única o una lista de etiquetas.
*   **`axis`**: Especifica si se eliminan filas o columnas:
    *   `axis=0` (o `axis='index'`) para eliminar filas (es el valor por defecto).
    *   `axis=1` (o `axis='columns'`) para eliminar columnas.
*   **`inplace`**: Un valor booleano (`True` o `False`). Si es `True`, la operación se realiza directamente sobre el DataFrame original y no devuelve nada. Si es `False` (por defecto), devuelve un nuevo DataFrame con los elementos eliminados y el original permanece sin cambios.
*   **`errors`**: Puede ser `'ignore'` o `'raise'` (por defecto). Si es `'ignore'`, suprime los errores si alguna de las etiquetas especificadas no se encuentra. Si es `'raise'`, lanzará un error si una etiqueta no existe.

```python
print('DataFrame original:')
print(df)

# Eliminar una sola fila por su índice (axis=0 es el valor por defecto)
# Esto devuelve una copia del DataFrame sin la fila 0
df_sin_fila_0 = df.drop(0)
print('\nDataFrame sin la fila 0:')
print(df_sin_fila_0.head())

# Eliminar múltiples filas por sus índices
df_sin_filas_1_y_3 = df.drop([1, 3])
print('\nDataFrame sin las filas 1 y 3:')
print(df_sin_filas_1_y_3.head())

# Eliminar una columna (usando axis=1)
df_sin_columna_sexo = df.drop('sexo', axis=1)
print("\nDataFrame sin la columna 'sexo':")
print(df_sin_columna_sexo.head())

# Eliminar múltiples columnas
df_sin_peso_y_altura = df.drop(['peso', 'altura'], axis=1)
print("\nDataFrame sin las columnas 'peso' y 'altura':")
print(df_sin_peso_y_altura.head())

# Eliminar filas modificando el DataFrame original (inplace=True)
# Ten cuidado con inplace=True, ya que modifica el DataFrame directamente.

df = pd.read_csv('https://raw.githubusercontent.com/asalber/manual-python/master/datos/colesteroles.csv', sep=';', decimal=',')

nueva_fila = pd.Series(['Carlos Rivas', 28, 'H', 89.0, 1.78, 245.0], index=['nombre','edad','sexo','peso','altura','colesterol'])
df = pd.concat([df, nueva_fila.to_frame().T], ignore_index=True)

print('\nDataFrame original antes de inplace=True:')
print(df.tail())

df.drop(14, inplace=True) # Eliminar la última fila (Carlos Rivas)
print('\nDataFrame después de eliminar la fila 14 con inplace=True:')
print(df.tail())
```

### Aplicar una función de agregación por grupos
Una vez dividido el DataFame en grupos, es posible aplicar funciones de agregación a cada grupo
mediante el siguiente método:

• `df.groupby(columnas).agg(funciones)` : Devuelve un DataFrame con el resultado de
aplicar las funciones de agregación de la lista `funciones` a cada uno de los DataFrames que
resultan de dividir el DataFrame según las `columnas` especificadas.

La función `agg` es muy flexible y puede recibir diferentes tipos de argumentos para `funciones`:

1.  **Una sola función:** Aplica la misma función a todas las columnas seleccionadas.
    *   Ejemplo: `df.groupby('sexo')[['edad', 'peso']].agg(np.mean)`

2.  **Una lista de funciones:** Aplica todas las funciones de la lista a cada columna seleccionada.
    *   Ejemplo: `df.groupby('sexo')[['edad', 'peso']].agg([np.mean, np.max])`

3.  **Un diccionario:** Permite especificar diferentes funciones o listas de funciones para columnas específicas.
    *   Ejemplo: `df.groupby('sexo').agg({'edad': 'mean', 'peso': ['min', 'max']})`

Las funciones de agregación pueden ser objetos de función (como `np.mean`, `np.min`, `len`) o sus nombres como cadenas de texto (como `'mean'`, `'min'`, `'count'`). Una función de agregación toma como argumento una serie de valores y devuelve un único valor.

Algunas de las funciones de agregación más comunes son:
• `np.min` (o `'min'`) : Devuelve el mínimo de una lista de valores.
• `np.max` (o `'max'`) : Devuelve el máximo de una lista de valores.
• `np.count_nonzero` (o `'count'`) : Devuelve el número de valores no nulos de una lista de valores.
• `np.sum` (o `'sum'`) : Devuelve la suma de una lista de valores.
• `np.mean` (o `'mean'`) : Devuelve la media de una lista de valores.
• `np.std` (o `'std'`) : Devuelve la desviación típica de una lista de valores.
• `len` (o `'size'`) : Devuelve el número total de elementos (incluyendo nulos).

```python
import numpy as np

# Ejemplo 1: Media de columnas numéricas agrupadas por 'sexo'
print('Media por sexo:')
display(df.groupby('sexo')[['edad', 'peso', 'altura', 'colesterol']].agg(np.mean))

# Ejemplo 2: Múltiples agregaciones para las mismas columnas
print('\nMínimo, máximo y media por sexo:')
display(df.groupby('sexo')[['edad', 'peso', 'altura', 'colesterol']].agg([np.min, np.max, np.mean]))

# Ejemplo 3: Agregaciones diferentes para columnas específicas usando un diccionario
print('\nAgregaciones personalizadas por sexo:')
display(df.groupby('sexo').agg({
    'edad': ['mean', 'min'],
    'peso': 'max',
    'altura': np.std,
    'colesterol': 'count' # Contar valores no nulos
}))
```

### Reestructurar un DataFrame
A menudo la disposición de los datos en un DataFrame no es la adecuada para su tratamiento y es
necesario reestructurar el DataFrame. Los datos que contiene un DataFrame pueden organizarse en
dos formatos: ancho y largo.
Formatos de un DataFrame

### Convertir un DataFrame a formato largo
Para convertir un DataFrame de formato ancho a formato largo (columnas a filas) se utiliza el siguiente
método:
• df.melt(id_vars=id-columnas, value_vars=columnas, var_name=nombre-
columnas, var_value=nombre-valores) : Devuelve el DataFrame que resulta de
convertir el DataFrame df de formato ancho a formato largo. Todas las columnas de lista
columnas se reestructuran en dos nuevas columnas con nombres nombre-columnas y
nombre-valores que contienen los nombres de las columnas originales y sus valores, res‑
pectivamente. Las columnas en la lista id-columnas se mantienen sin reestructurar. Si no se
pasa la lista columnas entonces se reestructuran todas las columnas excepto las columnas de
la lista id-columnas.

```python
datos = {'nombre':['María', 'Luis', 'Carmen'],
         'edad':[18, 22, 20],
         'Matemáticas':[8.5, 7, 3.5],
         'Economía':[8, 6.5, 5],
         'Programación':[6.5, 4, 9]}
df = pd.DataFrame(datos)
df
```

```python
df1 = df.melt(id_vars=['nombre', 'edad'], var_name='asignatura',
value_name='nota')
print(df1)
```

### Convertir un DataFrame a formato ancho
Para convertir un DataFrame de formato largo a formato ancho (filas a columnas) se utiliza el siguiente
método:
• df.pivot(index=filas, columns=columna, values=valores) : Devuelve el DataFra‑
me que resulta de convertir el DataFrame df de formato largo a formato ancho. Se crean tantas
columnas nuevas como valores distintos haya en la columna columna. Los nombres de estas
nuevas columnas son los valores de la columna columna mientras que sus valores se toman de la columna valores. Los nombres del índice del nuevo DataFrame se toman de los valores de
la columna filas.

```python
print(df1.pivot(index='nombre', columns='asignatura', values='nota'))
```

### Exportar resultados

Una vez que hayas procesado tus datos y obtenido los resultados deseados en un DataFrame, a menudo querrás guardarlos en un archivo para su uso posterior o para compartir. pandas ofrece métodos sencillos para exportar DataFrames a varios formatos, siendo CSV y Excel los más comunes.

**`df.to_csv(path_or_buf, sep, index)`**: Exporta el DataFrame `df` a un archivo CSV.
* `path_or_buf`: La ruta y nombre del archivo donde se guardará el DataFrame.
* `sep`: El separador de valores (por defecto es coma ',').
* `index`: Un valor booleano que indica si se debe incluir el índice del DataFrame en el archivo (por defecto es `True`).


```python
# Exportar el DataFrame 'df' a un archivo CSV
# Se excluye el índice del DataFrame en el archivo CSV.
df.to_csv('colesteroles_procesado.csv', index=False, sep=';')

print("DataFrame 'df' exportado a 'colesteroles_procesado.csv'")

# verificación
df_exported = pd.read_csv('colesteroles_procesado.csv', sep=';')
print("\nPrimeras 5 filas del DataFrame exportado:")
display(df_exported.head())
```
