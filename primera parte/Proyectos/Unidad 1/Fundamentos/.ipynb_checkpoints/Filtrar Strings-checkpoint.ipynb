# Filtrar Strings

A menudo, antes de formatear una frase, los datos vienen con "ruido" (espacios extra, mayúsculas desordenadas o caracteres no deseados). Python ofrece métodos integrados para solucionar esto rápidamente.

```python
dato_crudo = "   mi univerSiDAD es ULP y estoY apRendiEnDo PYtHon     "
print(f"Frase original: '{dato_crudo}'")
print(f"Longitud de dato original: {len(dato_crudo)}")
```

## 1. Eliminando espacios en blanco (strip)
Cuando recibimos datos de un usuario o una base de datos, es común encontrar espacios accidentales al inicio o al final.

```python
# strip() elimina espacios en ambos extremos y lower pasa el texto a minuscula
dato_limpio=dato_crudo.strip()
print(f"Frase limpia: '{dato_limpio}'")
print(f"Longitud de dato limpiol: {len(dato_limpio)}")
```

## 2. Control de Mayúsculas y Minúsculas
Para que nuestro formato sea consistente, podemos usar:
* lower(): Todo a minúsculas.
* upper(): Todo a mayúsculas.
* title(): Primera letra de cada palabra en mayúscula.
* capitalize(): Solo la primera letra de la frase en mayúscula.

```python
# strip() elimina espacios en ambos extremos y lower pasa el texto a minuscula
dato_limpio=dato_limpio.title()
print(f"Minusculas: '{dato_limpio.lower()}'")
print(f"Mayusculas: {dato_limpio.upper()}")
```

```python
print(f"Dato crudo: '{dato_limpio }'")
print(f"Limpio: '{dato_limpio}'")
print(f"Cantidad de palabras: {dato_limpio.count(" ") + 1}")
```

## 3. Reemplazo de caracteres (replace)
   Ten en cuenta que a dato_limpio le estamos asignando una nueva cadena de datos resultado de replace. Los strings son inmutables

```python
dato_limpio = dato_limpio.replace("Ulp", "La Punta")

print(f"Dato limpio: {dato_limpio}")
```

## 4. El poder del Encadenamiento (Method Chaining)
Lo mejor de estos métodos es que puedes usarlos uno tras otro en una sola línea. Esto mantiene tu código limpio y legible.

```python
# Dato de entrada con muchos problemas
entrada_sucia = "   uNiVeRsIdAd dE La pUnTa   "

# Limpiamos espacios, pasamos a minúsculas y luego corregimos formato
# El orden importa: primero quitamos espacios, luego aplicamos formato
entrada_perfecta = entrada_sucia.strip().lower().title()

print(f"Original: '{entrada_sucia}'")
print(f"Resultado: '{entrada_perfecta}'")
```

## 5. Segmentación de Strings (split)
El método split() divide un string en una lista de subcadenas basándose en un separador (por defecto, el espacio en blanco).

```python
correo = "alumno@ulp.edu.ar"
partes = correo.split("@")

print(f"Usuario: {partes[0]}")
print(f"Dominio: {partes[1]}")
```

## 6. La unión de strings (join)
join() toma una lista de strings y los une en uno solo.

**Nota importante:** A diferencia de otros métodos, join() se llama desde el separador que queremos usar, no desde la lista.

```python
tecnologias = ["Python", "Pandas", "Data Science"]

# Unir con una coma y un espacio
etiquetas = ", ".join(tecnologias)
# Unir con un guion para crear un nombre de archivo por ejemplo
nombre_archivo = "-".join(tecnologias).lower()
print(f"Archivo: {nombre_archivo}.csv") 
```
