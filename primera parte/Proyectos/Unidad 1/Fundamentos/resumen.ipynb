# 📘 MACHETE DE PYTHON - Comandos Esenciales

Consulta rápida de los comandos más importantes

## 1. TIPOS DE DATOS BÁSICOS

```python
# Enteros (int)
x = 10
print(type(x))  # int

# Decimales (float)
y = 3.14
print(type(y))  # float

# Booleanos (bool)
verdad = True
falso = False
print(type(verdad))  # bool

# Strings (str)
texto = "Hola mundo"
print(type(texto))  # str
```

## 2. OPERACIONES MATEMÁTICAS

```python
suma = 10 + 5        # 15
resta = 10 - 5       # 5
multi = 10 * 5       # 50
division = 10 / 3    # 3.333...
division_entera = 10 // 3   # 3
modulo = 10 % 3      # 1
potencia = 2 ** 3    # 8
raiz = 25 ** 0.5     # 5.0

print(f"Suma: {suma}")
print(f"Resta: {resta}")
print(f"Multiplicación: {multi}")
print(f"División: {division}")
print(f"División entera: {division_entera}")
print(f"Módulo: {modulo}")
print(f"Potencia: {potencia}")
print(f"Raíz: {raiz}")
```

## 3. OPERADORES DE COMPARACIÓN

```python
x = 5
y = 3

print(f"x == y: {x == y}")   # False (igual)
print(f"x != y: {x != y}")   # True (distinto)
print(f"x > y: {x > y}")     # True (mayor)
print(f"x < y: {x < y}")     # False (menor)
print(f"x >= y: {x >= y}")   # True (mayor o igual)
print(f"x <= y: {x <= y}")   # False (menor o igual)
```

## 4. OPERADORES BOOLEANOS

```python
print(f"True and True: {True and True}")   # True
print(f"True and False: {True and False}") # False
print(f"True or False: {True or False}")   # True
print(f"not True: {not True}")             # False
print(f"not False: {not False}")           # True
```

## 5. MÉTODOS PARA STRINGS

```python
texto = "   hOlA MUnDo   "

print(f"Original: '{texto}'")
print(f"strip(): '{texto.strip()}'")           # elimina espacios
print(f"lower(): '{texto.lower()}'")           # a minúsculas
print(f"upper(): '{texto.upper()}'")           # a mayúsculas
print(f"title(): '{texto.title()}'")           # primeras letras mayúsculas
print(f"capitalize(): '{texto.capitalize()}'") # primera letra mayúscula
print(f"replace(): '{texto.replace('Mundo', 'Python')}'")  # reemplazar
print(f"split(): {texto.split()}")             # dividir en lista

# join() - unir lista en string
lista = ["a", "b", "c"]
print(f"join(): {', '.join(lista)}")
```

## 6. INDEXACIÓN Y SLICING

```python
palabra = "python"
#          012345

print(f"palabra[0]: {palabra[0]}")     # 'p' (primer carácter)
print(f"palabra[-1]: {palabra[-1]}")   # 'n' (último carácter)
print(f"palabra[0:3]: {palabra[0:3]}") # 'pyt' (índice 0 al 2)
print(f"palabra[:3]: {palabra[:3]}")   # 'pyt' (desde inicio)
print(f"palabra[3:]: {palabra[3:]}")   # 'hon' (hasta final)
print(f"palabra[::2]: {palabra[::2]}") # 'pto' (saltando de 2 en 2)
print(f"palabra[::3]: {palabra[::3]}") # 'ph' (saltando de 3 en 3)
print(f"palabra[::-1]: {palabra[::-1]}") # 'nohtyp' (invertido)
```

## 7. ESTRUCTURAS DE CONTROL

```python
# if - elif - else
edad = 18
if edad >= 18:
    print("Adulto")
elif edad >= 13:
    print("Adolescente")
else:
    print("Niño")

# Condiciones compuestas
x = 7
if x > 5 and x < 10:
    print("x está entre 5 y 10")
```

## 8. LOOPS

```python
# Loop FOR
print("For con string:")
for letra in "python":
    print(f"  {letra}")

print("\nFor con lista:")
for num in [1, 2, 3]:
    print(f"  {num}")

# Loop WHILE
print("\nWhile:")
contador = 0
while contador < 3:
    print(f"  {contador}")
    contador += 1

# Break y Continue
print("\nBreak y Continue:")
for i in range(10):
    if i == 3:
        continue  # salta esta iteración
    if i == 7:
        break     # termina el loop
    print(f"  {i}")
```

## 9. FUNCIÓN RANGE()

```python
print("range(5):")
for i in range(5):
    print(f"  {i}")  # 0,1,2,3,4

print("\nrange(2, 7):")
for i in range(2, 7):
    print(f"  {i}")  # 2,3,4,5,6

print("\nrange(1, 10, 2):")
for i in range(1, 10, 2):
    print(f"  {i}")  # 1,3,5,7,9
```

## 10. FUNCIONES ÚTILES

```python
# Conversiones
print(f"int('123'): {int('123')}")      # 123
print(f"str(123): {str(123)}")          # "123"

# Longitud
print(f"len('python'): {len('python')}")  # 6
print(f"len([1,2,3]): {len([1,2,3])}")    # 3

# input() - descomentar para probar
# nombre = input("Dime tu nombre: ")
# print(f"Hola {nombre}")

# f-strings
nombre = "Juan"
print(f"Hola {nombre}")
```

## 11. EJEMPLOS COMBINADOS

```python
# Limpieza de texto encadenada
texto_sucio = "   UnIverSiDAD De La PuNTA   "
texto_limpio = texto_sucio.strip().lower().title()
print(f"Limpieza: '{texto_limpio}'")  # "Universidad De La Punta"

# Dividir y procesar
correo = "alumno@ulp.edu.ar"
usuario, dominio = correo.split("@")
print(f"Usuario: {usuario}")
print(f"Dominio: {dominio}")

# Loop con condicional
numeros = [1, 5, 8, 12, 3, 7]
print("Números mayores que 5:")
for n in numeros:
    if n > 5:
        print(f"  {n}")
```
