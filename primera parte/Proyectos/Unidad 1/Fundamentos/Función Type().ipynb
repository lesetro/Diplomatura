# Función Type()

En esta lección te voy a enseñar una función incorporada en Python, a la que la mayoría de los instructores no le dan la gran importancia que tiene, y es la función **type()**.

Las funciones son métodos que python tiene definidos para trabajar con los objetos. Podríamos decir que las funciones sirven para que los objetos de Python hagan cosas.

¿Y qué cosas hace la función type? La función type hace que los objetos nos digan qué tipo de datos son.

```python
a = 1
b = "1"
```

```python
a
```

```python
b
```

```python
print(b)
```

```python
type(a)
```

```python
type(b)
```

type() Te permite:
1. Evitar errores de "incompatibilidad"
Muchos errores ocurren cuando intentas sumar cosas que no se llevan bien (como un número y un texto). type() les permite depurar su código al instante.

Ejemplo: No puedes sumar 5 + "10". Usar type() les ayuda a ver que uno es int y el otro es str.

2. Entender la naturaleza de los datos
En Python, todo es un objeto. Saber qué tipo de dato tienen entre manos les dice qué pueden hacer con él:

Si es una lista, pueden agregar elementos.

Si es un diccionario, pueden buscar llaves.

Si es un booleano, pueden usarlo en una decisión (if/else).

3. Autonomía al programar
Cuando usan librerías externas o reciben datos de una base de datos, no siempre saben qué están recibiendo. type() les da el poder de inspeccionar lo desconocido sin tener que adivinar.
