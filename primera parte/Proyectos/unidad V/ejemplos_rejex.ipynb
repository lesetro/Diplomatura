```python
#cuando usamos la funcion findall buscamos substrings válidos dentro de un texto. No susamos anclas.
#patron para nombre - findall me devuelve una lista
import re
patron_nombre=r'[A-ZÁÉÍÓÚÑ][a-záéíóúñ]{1,}(?:[\s-][A-ZÁÉÍÓÚÑ][a-záéíóúñ]{1,})+'
x=re.findall(patron_nombre,texto)
print(x)
```

```python
#filtramos para mail
patron_mail=r'[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}'
x=re.findall(patron_mail,texto)
print(x)
```

```python
#filtramos para mail con search
linea='Cliente: Ana López - Tel: 351-456-7890 - Email: ana.lopez@mail.com'
linea1=''
patron_mail=r'[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}'
x=re.search(patron_mail,linea)
x=x.group() if x else None
print(x)
```

```python
#para validar con match - búsqueda literal, usamos anclas al inicio y final del patrón
patron_mail = r'^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,3}$'
email = "juan@gmail.com"
x=re.match(patron_mail, email)
x=x.group() if x else None
print(x)
```

```python
#filtramos telefonos
#patron_tel=r'\+54\s?9\s?|(\+1[\s-]?)?(\(?\d{2,4}\)?[\s-]?)?\d{3,4}[\s-]?\d{4}'
patron_tel = r'(?:\+54\s?9\s?|(?:\+1[\s-]?)?(?:\(?\d{2,4}\)?[\s-]?)?)\d{3,4}[\s-]?\d{4}'
x=re.findall(patron_tel,texto)
print(x)
```

```python
#Busca todas las coincidencias del patrón en el texto.
#No devuelve una lista de strings(como findall), sino un iterador con información extra  m.group(), m.start(), m.end()
for m in re.finditer(patron_tel, texto):
    print(f"coincide: {m.group()}, posicion de inicio: {m.start()}, posicion de fin: {m.end()}")
```

```python
#filtramos fechas
patron_fecha=r'(?:\d{2}[-/]\d{2}[-/]\d{4})|(?:\d{4}[-/]\d{2}[-/]\d{2})|(?:\d{1,2}\sde\s[A-Za-zÁÉÍÓÚáéíóúÑñ]+\sde\s\d{4})'
x=re.findall(patron_fecha,texto)
print(x)
```

```python
#filtramos direcciones web
patron_web=r'(?:https?:\/\/|ftp:\/\/)?(?:www\.)?[A-Za-z0-9]+\.[A-Za-z]{2,4}(?:[\/\w\.-]*)*\/?(?:\?[^\s]*)?'
x=re.findall(patron_web,texto)
print(x)
```
