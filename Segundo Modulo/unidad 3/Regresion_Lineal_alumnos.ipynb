#Ejemplo de Rendimiento Acádemico de los alumnos

1- Importamos las librerias

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn import linear_model
from sklearn.metrics import mean_absolute_error, mean_squared_error
```

2- Creamos un DataFrame

```python
df = pd.DataFrame()
df["Horas de estudio"] = [1, 2, 3, 4, 5]
df["Puntaje"] = [52, 56, 65, 70, 75]
```

3- Graficar  usando **matplotlib** para ver cómo se ve esto.

```python
plt.scatter(df['Horas de estudio'], df['Puntaje'], color='steelblue')
plt.xlabel('Horas de estudio')
plt.ylabel('Puntaje en el examen')
plt.title('Relación entre horas de estudio y puntaje obtenido')
plt.grid(True)
plt.show()
```

3- Definimos X e y

```python
# X debe ser 2D (doble corchete) para sklearn
X = df[['Horas de estudio']]
y = df['Puntaje']
print('X:')
print(X)
print('y:', y.values)
```

4- Creamos conjunto dse prueba

```python
test = pd.DataFrame()
test['Horas de estudio'] = [6, 7, 8]
test['Puntaje']          = [75, 80, 85]

X_test = test[['Horas de estudio']]
y_test = test['Puntaje']

print('Conjunto de prueba:')
print(test)
```

5- Creamos Modelo

```python
modelo = linear_model.LinearRegression()
print('Modelo creado:', modelo)
```

6- Entrenamos

```python
modelo.fit(X, y)
print('Modelo entrenado')
print(f'Pendiente (coeficiente): {modelo.coef_[0]:.4f}')
print(f'Intersección (bias):     {modelo.intercept_:.4f}')
print()
print('Ecuación: Puntaje = {:.2f} * Horas + {:.2f}'.format(
    modelo.coef_[0], modelo.intercept_))
```

7- Predecir  y Comparamos



```python
y_pred = modelo.predict(X_test)

comparacion = pd.DataFrame({
    'Horas de estudio': X_test['Horas de estudio'].values,
    'Puntaje real'    : y_test.values,
    'Puntaje predicho': y_pred.round(2)
})
print(comparacion)
```

8- Visualizar

```python
# Scatter de los datos reales
plt.scatter(X, y, color='blue', label='Datos reales')
plt.xlabel('Horas de estudio')
plt.ylabel('Puntaje')
plt.title('Datos reales de entrenamiento')
plt.grid(True)
plt.legend()
plt.show()

# Línea de regresión sobre los datos de prueba
plt.plot(X_test, modelo.predict(X_test), color='green', label='Línea de regresión')
plt.xlabel('Horas de estudio')
plt.ylabel('Puntaje')
plt.title('Línea de regresión')
plt.grid(True)
plt.legend()
plt.show()
```

Y si juntamos nuestros dos gráficos en un solo axes, esto es lo que obtenemos:

```python
# Datos reales
plt.figure(figsize=(8, 5))
plt.scatter(X, y, color='blue', label='Datos reales')

# Línea de regresión
plt.plot(X_test, modelo.predict(X_test), color='green', label='Línea de regresión')

# Nueva predicción
nueva_hora = 9
nueva_muestra = pd.DataFrame([[nueva_hora]], columns=["Horas de estudio"])
nuevo_puntaje = modelo.predict(nueva_muestra)[0]
plt.scatter([nueva_hora], [nuevo_puntaje], color='red', marker='x', s=100, label='Predicción (8 horas)')

# Estética
plt.title("Relación entre horas de estudio y puntaje obtenido")
plt.xlabel("Horas de estudio")
plt.ylabel("Puntaje en el examen")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show();
```

### Evaluación del modelo
Además del coeficiente R² que devuelve `.score()`, es útil evaluar el modelo utilizando métricas adicionales como:

- **MAE (Mean Absolute Error)**: error promedio absoluto entre los valores reales y predichos.
- **MSE (Mean Squared Error)**: error cuadrático medio.
- **RMSE (Root Mean Squared Error)**: raíz del error cuadrático medio.


Las **métricas de error** nos dan una idea de qué tan bien se ajusta el modelo de regresión a los datos reales:

-**MAE** (Error Absoluto Medio) de 0.88 significa que, en promedio, el modelo se equivoca en 0.88 puntos al estimar el puntaje de los estudiantes. Es una medida fácil de interpretar porque mantiene la misma unidad que la variable objetivo.

-**MSE** (Error Cuadrático Medio) de 1.04 penaliza más los errores grandes (porque eleva las diferencias al cuadrado). Aunque no es tan fácil de interpretar directamente, sirve para comparar modelos: cuanto menor, mejor.

-**RMSE** (Raíz del Error Cuadrático Medio) de 1.02 también está en la misma escala que el puntaje, por lo que indica que los errores promedio están apenas por encima de 1 punto. Es útil cuando querés tener una noción de la dispersión típica del error.

En este contexto, los errores son bajos y bastante cercanos entre sí, lo cual indica que el modelo logra un ajuste razonable para un problema sencillo y con pocos datos.

9. Calcular el Error

```python
mae  = mean_absolute_error(y_test, y_pred)
mse  = mean_squared_error(y_test, y_pred)
rmse = mse ** 0.5

print(f'MAE  (Error Absoluto Medio)      : {mae:.4f}')
print(f'MSE  (Error Cuadrático Medio)    : {mse:.4f}')
print(f'RMSE (Raíz del Error Cuadrático) : {rmse:.4f}')
```

# 10 - Puntaje del modelo (R²)
El coeficiente de determinación **R²** indica qué proporción de la variabilidad del puntaje puede explicarse por la variable 'horas de estudio'. Su valor está entre 0 y 1, donde 1 significa que el modelo explica perfectamente los datos.

```python
r2 = modelo.score(X_test, y_test)
print(f'R² = {r2:.4f}')
print()
print(f'El modelo explica el {r2*100:.1f}% de la variabilidad del puntaje.')
if r2 >= 0.9:
    print('Excelente ajuste.')
elif r2 >= 0.7:
    print('Buen ajuste.')
else:
    print('Ajuste moderado — considerar más datos o features.')
```
