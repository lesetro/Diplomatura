## Parte 2 — Media, mediana y moda



### Ejercicio 2.1 — Cálculo usando la definición

Vamos a calcular la media, mediana y moda del salario 'a mano' para entender mejor cómo se construyen estas métricas.



```
# Media
x = (x1 + x2 + ... + xn) / n-1
```





### Ejercicio 2.2 — Cálculo de las tres medidas sobre `edad`

````python
# Con statistics:
media_edad   = statistics.mean(df['edad'])
mediana_edad = statistics.median(df['edad'])
moda_edad    = statistics.mode(df['edad'])

# Con pandas:
media_pd   = df['edad'].mean()
mediana_pd = df['edad'].median()
moda_pd    = df['edad'].mode().values

print('=== EDAD ===')
print(f'  Media   (statistics): {media_edad}')
print(f'  Mediana (statistics): {mediana_edad}')
print(f'  Moda    (statistics): {moda_edad}')
print(f'  Media   (pandas):     {media_pd}')
print(f'  Mediana (pandas):     {mediana_pd}')
print(f'  Moda    (pandas):     {moda_pd}')
````

**Interpretación:**  
Las tres medidas dan valores similares (Media ≈ 33.8, Mediana ≈ 31), lo que indica que la distribución de edades **no está muy sesgada**. Cuando media ≈ mediana, los datos son aproximadamente simétricos.

> **Respuesta:** Las edades se distribuyen de manera equilibrada; no hay valores extremos que distorsionen la media significativamente.


### Ejercicio 2.3 — Cálculo sobre `salario`

````python
print('=== SALARIO ===')
print(f"  Media   : {df['salario'].mean():.2f}")
print(f"  Mediana : {df['salario'].median():.2f}")
print(f"  Moda    : {df['salario'].mode().values}")
````

**Interpretación:**  
La **media** del salario es mucho mayor que la **mediana**, lo que delata la presencia de Héctor ($320.000). La mediana representa mejor el salario típico porque es robusta ante outliers: el 50% de los empleados gana menos que ese valor, sin importar cuánto gana Héctor.  

> **Respuesta:** Usaría la **mediana** para representar el salario típico, ya que la media se ve inflada por el outlier de Héctor.


### Ejercicio 2.4 — Moda en variables categóricas

````python
print('Moda de departamento    :', df['departamento'].mode().values)
print('Moda de nivel_educativo :', df['nivel_educativo'].mode().values)
print()
print('Para variables NOMINALES (departamento): solo la MODA tiene sentido.')
print('Para variables ORDINALES (nivel_educativo): MODA y MEDIANA son válidas,')
print('pero NO la media (promedio de Secundario y Posgrado no tiene significado).')
````

**Interpretación:**  
- **Variables nominales** (departamento): solo la **moda** tiene sentido. No existe orden entre categorías.
- **Variables ordinales** (nivel_educativo, satisfaccion): son válidas la **moda** y la **mediana**. La media no aplica porque los intervalos entre categorías no son iguales.

> **Respuesta:** Nominal → solo moda. Ordinal → moda y mediana.

## Parte 3 — Efecto de los outliers

Héctor tiene un salario de $320.000, muy por encima del resto del equipo.

### Ejercicio 3.1 — Comparar con y sin outlier


````python
# Con Héctor (ya calculado arriba)
media_con   = df['salario'].mean()
mediana_con = df['salario'].median()

# Sin Héctor
df_sin = df[df['nombre'] != 'Héctor']
media_sin   = df_sin['salario'].mean()
mediana_sin = df_sin['salario'].median()

print(f"{'':22s} {'CON Héctor':>12s} {'SIN Héctor':>12s}")
print(f"{'Media':22s} {media_con:>12.2f} {media_sin:>12.2f}")
print(f"{'Mediana':22s} {mediana_con:>12.2f} {mediana_sin:>12.2f}")
````

### Ejercicio 3.2 — Percentiles y Cuartiles

Los percentiles nos permiten entender la distribución de los datos y dónde se ubica un valor particular dentro de esa distribución. Los cuartiles son percentiles específicos (25, 50 y 75).

Calculá los percentiles 25, 50, 75 y 90 del 'salario' para entender mejor cómo se distribuyen los ingresos en la empresa.

````python
percentiles = df['salario'].quantile([0.25, 0.50, 0.75, 0.90])
print('Percentiles de salario:')
print(f"  P25 (Q1) : {percentiles[0.25]:>10,.0f}")
print(f"  P50 (Q2) : {percentiles[0.50]:>10,.0f}")
print(f"  P75 (Q3) : {percentiles[0.75]:>10,.0f}")
print(f"  P90      : {percentiles[0.90]:>10,.0f}")
````

**Interpretación:**

1. **Percentil 25 (Q1):** el 25% de los empleados gana igual o menos que ese valor. **Percentil 75 (Q3):** el 75% de los empleados gana igual o menos que ese valor; el 25% restante gana más.

2. Los percentiles revelan cómo se reparten los datos en todo el rango, no solo el centro. Permiten detectar asimetría y outliers mejor que la media sola. Por ejemplo, si P90 es muchísimo mayor que P75, hay una cola derecha muy larga.

> **Respuesta:** El P90 altísimo (Héctor) confirma la asimetría del salario. El P50 representa mejor el salario típico.


**Conclusión final:**

1. La **media** se vio mucho más afectada que la mediana por la presencia de Héctor.

2. Usaría la **mediana** para comunicar el salario representativo a los empleados, porque no se distorsiona por valores extremos.

3. Situaciones reales donde esto genera errores: empresas que reportan el 'salario promedio' inflado por directivos de altos sueldos, o inmobiliarias que publican el 'precio promedio' de propiedades siendo distorsionado por unos pocos inmuebles de lujo. Los empleados/compradores se formarían expectativas erróneas sobre lo que es 'normal'.

> **Recomendación:** Siempre reportar **media Y mediana** e indicar si hay outliers relevantes.
