### 2. Acceso y Operaciones con Tensores

Dada la definición de un tensor de matriz:
`t = tf.constant([[10., 20., 30.], [40., 50., 60.]])`

```python
import tensorflow as tf
t = tf.constant([[10., 20., 30.], [40., 50., 60.]])
```

**Ejercicio 2.1:** Acceder a la primera fila del tensor `t`.

```python
# Acceso a la primera fila del tensor
print("t =\n", t.numpy(), "\n")

primera_fila = t[0]          # equivale a t[0, :]  ->  "fila 0, todas las columnas"

print("t[0] ->", primera_fila.numpy())
print("shape:", primera_fila.shape, "-> quedó un vector de 3 elementos (rank 1)")
```

**Ejercicio 2.2:** Acceder a la última columna del tensor `t`.

```python
# Acceso a la última columna del tensor
ultima_columna = t[:, -1]    # ":" = todas las filas | "-1" = la última columna

print("t[:, -1] ->", ultima_columna.numpy())
print("shape   :", ultima_columna.shape)

# El -1 cuenta desde el final, así que sirve sin saber cuántas columnas hay.
# Es lo mismo que escribir t[:, 2] en este caso:
print("\n¿t[:, -1] == t[:, 2]? ->", bool(tf.reduce_all(t[:, -1] == t[:, 2])))
```

**Ejercicio 2.3:** Sumar el escalar `5.0` al tensor `t`.

```python
# Suma de un tensor y un escalar
suma_escalar = t + 5.0       # equivale a tf.add(t, 5.0)

print("t =\n", t.numpy(), "\n")
print("t + 5.0 =\n", suma_escalar.numpy())

# El 5.0 se suma a CADA elemento: esto es "broadcasting".
# TensorFlow estira el escalar hasta la forma (2, 3) sin gastar memoria extra.
```

**Ejercicio 2.4:** Sumar el tensor `t` consigo mismo (`t + t`).

```python
# Suma de dos tensores
suma_tensores = t + t        # equivale a tf.add(t, t)

print("t =\n", t.numpy(), "\n")
print("t + t =\n", suma_tensores.numpy())

# Acá la suma es elemento a elemento (posición por posición).
# Ambos tensores tienen la misma shape (2, 3), así que encajan directamente.
print("\nshape del resultado:", suma_tensores.shape)
```

**Ejercicio 2.5:** Calcular la raíz cuadrada de los elementos de un tensor constante `a = tf.constant([4.0, 9.0, 16.0, 25.0])` usando `tf.sqrt`.

```python
# Raíz cuadrada de un tensor
a = tf.constant([4.0, 9.0, 16.0, 25.0])

raiz = tf.sqrt(a)

print("a         =", a.numpy())
print("tf.sqrt(a) =", raiz.numpy())

# tf.sqrt se aplica a cada elemento por separado.
# Requiere un tensor de punto flotante: con tf.constant([4, 9]) (int32) daría error.
```

### 3. Conversión de Tipos

Una de las cosas relevantes a tener en cuenta cuando se utiliza TensorFlow es que **no realiza conversión implícita de tipos** al realizar una operación . Esto quiere decir que si no se utilizan valores del mismo tipo en la operación, se producirá una excepción .

**Se pide realizar las siguientes pruebas:**
1. Crear un tensor entero `a = tf.constant([1, 2, 3])` y un tensor flotante `b = tf.constant([1.0, 2.0, 3.0])`, y verificar sus `dtype`.
2. Intentar sumar `a + b` para observar el error `TypeError`.
3. Convertir el tensor `a` a flotante utilizando `tf.cast(a, tf.float32)` y realizar la suma correctamente.

```python
# 1. Crear tensores de distinto dtype e inspeccionarlos
a = tf.constant([1, 2, 3])          # sin punto decimal  -> int32
b = tf.constant([1.0, 2.0, 3.0])    # con punto decimal  -> float32

print("a =", a.numpy(), " dtype:", a.dtype)
print("b =", b.numpy(), " dtype:", b.dtype)

print("\n¿Son del mismo tipo? ->", a.dtype == b.dtype)
```

```python
# 2. Intentar sumar 'a' y 'b' (comprobar el error)
try:
    resultado = a + b
    print(resultado)
except Exception as e:
    print("❌ La operación falló, como era de esperar.")
    print("Tipo de excepción:", type(e).__name__)
    print("Mensaje:", str(e).split("\n")[0])

# TensorFlow NO hace conversión implícita de tipos: no convierte int32 a float32
# por su cuenta. Prefiere fallar antes que cambiar la precisión sin avisar.
#
# Nota: el enunciado menciona un TypeError. En TensorFlow 2 la excepción concreta
# es InvalidArgumentError; el concepto que se evalúa es el mismo.
```

```python
# 3. Usar tf.cast para convertir el tipo de dato y realizar la suma
a_float = tf.cast(a, tf.float32)     # int32 -> float32

print("a original  :", a.numpy(), " dtype:", a.dtype)
print("a convertido:", a_float.numpy(), " dtype:", a_float.dtype)

suma = a_float + b
print("\n✅ a_float + b =", suma.numpy(), " dtype:", suma.dtype)

# La alternativa es fijar el tipo desde el principio:
a_bis = tf.constant([1, 2, 3], dtype=tf.float32)
print("\nMismo resultado creándolo ya como float32:", (a_bis + b).numpy())
```

### 4. Tensores Variables y Mutabilidad

Hasta ahora, las estructuras de TensorFlow que hemos utilizado para almacenar valores (`tf.constant`) son **inmutables**, esto quiere decir que no podemos realizar modificaciones una vez que se han creado .

Probar que `t = tf.constant([10.0, 20.0, 30.0])` e intentar `t[0] = 99.0` produce una excepción .

```python
# Probar la inmutabilidad de tf.constant
t_const = tf.constant([10.0, 20.0, 30.0])
print("Tensor original:", t_const.numpy())

try:
    t_const[0] = 99.0
    print(t_const)
except Exception as e:
    print("\n❌ No se puede modificar un tf.constant.")
    print("Tipo de excepción:", type(e).__name__)
    print("Mensaje:", e)

# Un tf.constant es INMUTABLE: una vez creado, su contenido no se toca.
# Sí se pueden crear tensores nuevos a partir de él (t + 1 devuelve otro tensor),
# pero el original queda intacto:
print("\nDespués de t_const + 1, el original sigue igual:", (t_const + 1).numpy(), "vs", t_const.numpy())
```

Para modificar valores, debemos utilizar **tensores variables** (`tf.Variable`) .

**Ejercicio:** Executar y probar el código correspondiente para:
1. Crear `v = tf.Variable([100.0, 200.0, 300.0])`.
2. Imprimir el tensor original usando `v.value()`.
3. Modificar el elemento del índice `1` cambiando su valor a `250.0` usando `.assign()`.
4. Imprimir el tensor modificado.

```python
# Uso de tensores variables (tf.Variable)

# 1. Crear la variable
v = tf.Variable([100.0, 200.0, 300.0])

# 2. Imprimir el tensor original con .value()
print("1) Variable creada:")
print(v, "\n")
print("2) v.value() ->", v.value().numpy())

# 3. Modificar el elemento del índice 1, cambiándolo a 250.0
v[1].assign(250.0)

# 4. Imprimir el tensor modificado
print("\n3) Tras v[1].assign(250.0):")
print("4) v.value() ->", v.value().numpy())

# A diferencia de tf.constant, una tf.Variable SÍ es mutable.
# Por eso es el contenedor que TensorFlow usa para los pesos y bias de una red:
# durante el entrenamiento esos valores se actualizan en cada paso.

print("\n--- Otros métodos útiles ---")
v.assign([1.0, 2.0, 3.0])                # reemplaza todo el contenido
print("v.assign([1,2,3])      ->", v.value().numpy())
v.assign_add([10.0, 10.0, 10.0])         # suma en el lugar
print("v.assign_add(+10)      ->", v.value().numpy())
v.assign_sub([1.0, 1.0, 1.0])            # resta en el lugar
print("v.assign_sub(-1)       ->", v.value().numpy())

# assign NO permite cambiar la shape ni el dtype:
try:
    v.assign([1.0, 2.0])                 # 2 elementos en lugar de 3
except Exception as e:
    print("\n❌ Cambiar la shape con assign falla:", type(e).__name__)
```

### 5. Cadenas de Texto en TensorFlow

TensorFlow soporta otros tipos de estructuras de datos como las cadenas de texto .

**Ejercicio:** Probar la creación de tensores de cadenas de texto con:
- `tf.constant(b"machine learning")`
- `tf.constant("inteligencia artificial")`
- `tf.constant(["python", "tensorflow", "data science"])`

```python
# Tensores con cadenas de texto

# 1) Cadena ya expresada en bytes
s1 = tf.constant(b"machine learning")
print("s1:", s1)
print("   shape:", s1.shape, "| dtype:", s1.dtype, "| rank:", tf.rank(s1).numpy(), "\n")

# 2) Cadena de texto normal (Python la convierte a bytes automáticamente)
s2 = tf.constant("inteligencia artificial")
print("s2:", s2)
print("   shape:", s2.shape, "-> es un escalar: toda la frase es UN elemento\n")

# 3) Vector de cadenas
s3 = tf.constant(["python", "tensorflow", "data science"])
print("s3:", s3)
print("   shape:", s3.shape, "-> 3 elementos, sin importar el largo de cada palabra")

# Observaciones importantes:
# - El dtype siempre es tf.string y los valores se guardan como BYTES (el prefijo b'...').
# - Un tensor de texto NO tiene una dimensión por carácter: cada cadena es un elemento.

print("\n--- Volver a texto legible ---")
print([palabra.decode("utf-8") for palabra in s3.numpy()])

print("\n--- Algunas operaciones sobre strings ---")
print("Mayúsculas:", tf.strings.upper(s3).numpy())
print("Longitudes:", tf.strings.length(s3).numpy())
print("Split de s2:", tf.strings.split(s2).numpy())
```

### 6. Descripción de Atributos

**Ejercicio final:** Describe cada uno de los elementos/atributos principales que componen un tensor (ej. rango/dimensión, forma/shape, tipo de datos/dtype y dispositivo de ejecución) .

### Respuesta: los atributos que componen un tensor

**1. Rango / dimensión (*rank*)** — `tf.rank(t)` o `t.ndim`

Es la **cantidad de ejes** que tiene el tensor. Un número entero:

| Rank | Nombre | Ejemplo | Shape |
|---|---|---|---|
| 0 | Escalar | `42` | `()` |
| 1 | Vector | `[1., 2., 3.]` | `(3,)` |
| 2 | Matriz | `[[1,2,3], [4,5,6]]` | `(2, 3)` |
| 3 | Tensor 3D | Una imagen a color | `(alto, ancho, 3)` |
| 4 | Tensor 4D | Un lote de imágenes | `(n, alto, ancho, 3)` |

**2. Forma (*shape*)** — `t.shape`

Es el **tamaño de cada eje**, expresado como una tupla. Un tensor con shape `(2, 3)` tiene 2 filas y 3 columnas. La cantidad de números en la tupla es, justamente, el rank.

Es el atributo que más se consulta al depurar: la mayoría de los errores de TensorFlow son incompatibilidades de shape.

**3. Tipo de dato (*dtype*)** — `t.dtype`

El tipo de **todos** los elementos del tensor. Es **uniforme**: a diferencia de una lista de Python, un tensor no puede mezclar enteros con textos.

Los más habituales: `tf.float32` (el estándar en Deep Learning), `tf.float64`, `tf.int32`, `tf.string` y `tf.bool`.

>  TensorFlow **no convierte tipos automáticamente**. Para operar entre tensores de distinto dtype hay que usar `tf.cast()` — es lo que se practicó en el punto 3.

**4. Dispositivo de ejecución (*device*)** — `t.device`

Indica **dónde vive el tensor** en memoria y dónde se ejecutan sus operaciones: CPU o GPU. Devuelve algo como `/job:localhost/replica:0/task:0/device:CPU:0`.

Importa porque las GPUs, al tener miles de núcleos, aceleran enormemente las operaciones con matrices, que son la base del entrenamiento de redes neuronales. TensorFlow elige el dispositivo solo, pero se puede forzar con `with tf.device('/GPU:0'):`.

---

**Un quinto atributo que conviene conocer:** la **mutabilidad**. Un `tf.constant` es inmutable (no se puede modificar tras crearlo), mientras que un `tf.Variable` sí admite cambios mediante `.assign()`. Por eso los pesos de una red neuronal, que se actualizan en cada paso del entrenamiento, son siempre `tf.Variable`.

```python
# Demostración práctica de cada atributo

ejemplos = {
    "escalar (rank 0)": tf.constant(42),
    "vector  (rank 1)": tf.constant([1.0, 2.0, 3.0]),
    "matriz  (rank 2)": tf.constant([[10., 20., 30.], [40., 50., 60.]]),
    "cubo    (rank 3)": tf.zeros([2, 3, 4]),
    "texto            ": tf.constant(["python", "tensorflow"]),
}

print(f"{'TENSOR':<18} {'RANK':<6} {'SHAPE':<16} {'DTYPE':<10} {'SIZE':<6}")
print("-" * 62)
for nombre, x in ejemplos.items():
    print(f"{nombre:<18} {tf.rank(x).numpy():<6} {str(x.shape):<16} {x.dtype.name:<10} {tf.size(x).numpy():<6}")

print("\nDispositivo donde se ejecutan (device):")
print(" ", tf.constant([1.0]).device)
```
