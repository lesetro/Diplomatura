# Parte A — Teoría

## 1. El mapa: IA ⊃ ML ⊃ DL

Son **círculos concéntricos**, no tres cosas distintas. Cada uno está contenido en el anterior:

| Nivel | Qué es | Ejemplo |
|---|---|---|
| **IA** (Inteligencia Artificial) | Cualquier técnica que permita a una computadora imitar el comportamiento humano. Incluye cosas sin aprendizaje, como un sistema de reglas. | Un motor de ajedrez basado en reglas fijas |
| **ML** (Machine Learning) | Subconjunto de la IA: la máquina **aprende de los datos** sin ser programada explícitamente. | Filtro de spam, regresión lineal, árboles de decisión |
| **DL** (Deep Learning) | Subconjunto del ML: extrae patrones usando **redes neuronales profundas** (muchas capas). | Reconocimiento facial, traducción automática, autos autónomos |

> **La idea a retener:** todo Deep Learning es Machine Learning, y todo Machine Learning es IA. Pero no al revés.

**Definición de Deep Learning (de la presentación):**
> Técnica de IA que consiste en enseñar a las computadoras a procesar datos usando algoritmos inspirados en el cerebro humano. También se lo llama *aprendizaje neuronal profundo* o *redes neuronales profundas*: las computadoras adquieren el aprendizaje **a través de la observación**, de manera similar a las personas.

## 2. El cambio de paradigma: programación tradicional vs. Machine Learning

Esta es probablemente **la idea más importante de toda la unidad**. Lo que cambia es *qué es la entrada y qué es la salida*:

```
PROGRAMACIÓN TRADICIONAL
   Datos de entrada  ─┐
                      ├──►  Computadora  ──►  Datos de salida
   Programa (reglas) ─┘


MACHINE LEARNING
   Datos de entrada  ─┐
                      ├──►  Computadora  ──►  Programa (MODELO)
   Datos de salida   ─┘
```

- En la programación tradicional, **vos escribís las reglas** y la computadora produce resultados.
- En Machine Learning, **le das ejemplos de entradas y sus resultados correctos**, y la computadora *deduce las reglas sola*. Ese conjunto de reglas aprendidas es el **modelo**.

**El flujo completo de un proyecto de ML:**

```
   Dataset                 Generación                Predicción
(variables input   ──►     de un modelo      ──►     de nuevos outputs
 y output)                 (vía algoritmo)           a partir del modelo
```

> ⚠️ **Clave:** hace falta un **dataset con un volumen de datos considerable**. Sin datos suficientes no hay nada que aprender — este es el cuello de botella real de la mayoría de los proyectos.

## 3. Qué hace *profundo* al Deep Learning

La palabra "profundo" se refiere a la **cantidad de capas** de la red. Y cada capa aprende características de un nivel de abstracción más alto que la anterior.

El ejemplo clásico, reconocer una cara:

```
  Imagen  ──►  Low Level  ──►  Mid Level  ──►  High Level  ──►  "Persona"
              (bordes,        (ojo, nariz,     (la cara
               líneas,         oreja,           completa,
               manchas)        boca)            la identidad)
```

**Por qué esto fue una revolución:** antes del Deep Learning, un humano tenía que decidir *a mano* qué características medir (el ancho de la nariz, la distancia entre los ojos...). Eso se llamaba *feature engineering* y era lentísimo y frágil.

La red profunda **aprende sola qué características importan**, directamente desde los píxeles crudos. Vos le das imágenes y etiquetas; ella arma la jerarquía.

## 5. CPU vs. GPU

|  | **CPU** (serial) | **GPU** (paralelo) |
|---|---|---|
| Núcleos | Pocos, de alta potencia individual | **Miles**, de menor potencia individual |
| Memoria | Compartida | Dedicada |
| Ideal para | Tareas **secuenciales** | **Deep Learning y matrices** |
| Eficiencia en DL | ❌ Baja | ✅ Alta |

**Por qué la GPU gana tanto:** entrenar una red neuronal es, en el fondo, **multiplicar matrices gigantes una y otra vez**. Cada elemento del resultado se puede calcular *independientemente* de los demás. Eso es exactamente el tipo de trabajo que una GPU, con miles de núcleos, hace en paralelo — mientras la CPU los haría casi de a uno.

Es la diferencia entre 4 personas muy rápidas y 5.000 personas normales pintando una pared: para la pared, ganan las 5.000.

## 6. TensorFlow: qué es y cómo está armado

**TensorFlow** es la biblioteca de código abierto de **Google** para Machine Learning y Deep Learning. Es el **entorno oficial para las prácticas del curso**.

| Aspecto | Detalle |
|---|---|
| Desarrollador | Google |
| Multiplataforma | Windows, Linux, Mac, Android, iOS |
| Núcleo | **C++** (rendimiento) + **Python** (usabilidad) |
| Lenguajes | Python, C++, Java, Go |

**La arquitectura, de arriba hacia abajo:**

```
┌──────────────────────────────────────────────┐
│   Estimators  (API de alto nivel)            │  ← armás un modelo en pocas líneas
├───────────────┬──────────────┬───────────────┤
│    Layers     │   Datasets   │    Metrics    │  ← piezas para construir
├───────┬───────┼───────┬──────┴───────────────┤
│Python │  C++  │ Java  │        Go            │  ← lenguajes de acceso
├───────┴───────┴───────┴──────────────────────┤
│  TensorFlow Distributed Execution Engine     │  ← el motor: reparte el cálculo
└──────────────────────────────────────────────┘        entre CPU / GPU / varias máquinas
```

**Lo importante de este esquema:** vos escribís Python, pero el cálculo pesado lo ejecuta el motor en C++, distribuyéndolo sobre la CPU o la GPU. Por eso TensorFlow es rápido aunque lo manejes desde un lenguaje interpretado.

*Nota práctica:* la lámina menciona **Estimators** como API de alto nivel. En TensorFlow 2 esa API quedó **obsoleta** y su lugar lo ocupa **Keras** (`tf.keras`), que es lo que vas a usar cuando llegues a armar modelos.

# Parte B — El tensor en la práctica

## 7. Qué es un tensor

> **Definición:** un contenedor **multidimensional** de datos (similar a un `ndarray` de NumPy), **inmutable** y **acelerado por GPU**.

El tensor es **el elemento fundamental de TensorFlow** — de ahí el nombre de la biblioteca: *Tensor* + *Flow* (tensores que fluyen a través de un grafo de operaciones).

**Diferencias con un `ndarray` de NumPy:**

| | NumPy `ndarray` | TensorFlow `Tensor` |
|---|---|---|
| Multidimensional | ✅ | ✅ |
| Corre en GPU | ❌ | ✅ |
| Mutable | ✅ (`a[0] = 5` funciona) | ❌ (es inmutable) |
| Diferenciación automática | ❌ | ✅ (clave para entrenar redes) |
| Escalares sueltos | Los trata como array 0-D | Los admite naturalmente |

**El "rank" (rango) es la cantidad de dimensiones:**

| Rank | Nombre | Shape | Ejemplo | En el mundo real |
|---|---|---|---|---|
| 0 | **Escalar** | `()` | `15` | Un precio, una temperatura |
| 1 | **Vector** | `(3,)` | `[1.0, 2.0, 3.0]` | Las 4 características de una flor de IRIS |
| 2 | **Matriz** | `(2, 3)` | `[[1,2,3], [4,5,6]]` | Una tabla, una imagen en escala de grises |
| 3 | **Tensor 3D** | `(28, 28, 3)` | — | Una imagen a color (alto × ancho × RGB) |
| 4 | **Tensor 4D** | `(32, 28, 28, 3)` | — | Un **lote** de 32 imágenes a color |

> Cuando entrenás una red con imágenes, tus datos son literalmente un tensor de rank 4. Por eso importa entender esto ahora.

````python
import tensorflow as tf
import numpy as np

print("Versión de TensorFlow:", tf.__version__)
print("¿GPU disponible?     :", tf.config.list_physical_devices('GPU') or "No (se usará CPU)")
````

````python
# Los cinco ranks, uno por uno

escalar = tf.constant(15)                                   # rank 0
vector  = tf.constant([1.0, 2.0, 3.0])                      # rank 1
matriz  = tf.constant([[1, 2, 3], [4, 5, 6]])               # rank 2
cubo    = tf.zeros([2, 3, 4])                               # rank 3
lote    = tf.zeros([32, 28, 28, 3])                         # rank 4 (32 imágenes RGB de 28x28)

for nombre, t in [("escalar", escalar), ("vector", vector), ("matriz", matriz),
                  ("cubo 3D", cubo), ("lote 4D", lote)]:
    print(f"{nombre:9s} -> rank={tf.rank(t).numpy()}  shape={str(t.shape):18s} dtype={t.dtype.name}")
````

## 8. Los 4 atributos de un tensor

Este es justo **el ejercicio final del TP**, así que vale la pena tenerlo claro:

| Atributo | Cómo se consulta | Qué es |
|---|---|---|
| **Rank / dimensión** | `tf.rank(t)` o `t.ndim` | Cuántos ejes (dimensiones) tiene. Un número entero. |
| **Shape / forma** | `t.shape` | El tamaño de cada eje. Una tupla: `(2, 3)` = 2 filas, 3 columnas. |
| **Dtype / tipo** | `t.dtype` | El tipo de dato de **todos** los elementos: `float32`, `int32`, `string`... Es **uniforme**: un tensor no mezcla tipos. |
| **Device / dispositivo** | `t.device` | Dónde vive el tensor en memoria: CPU o GPU. |

Extras útiles: `tf.size(t)` (cantidad total de elementos) y `t.numpy()` (convertirlo a NumPy).

````python
t = tf.constant([[5., 10., 15.],
                 [20., 25., 30.]])

print("Tensor:")
print(t.numpy(), "\n")

print("rank (nº de dimensiones):", tf.rank(t).numpy())   # 2
print("shape (forma)           :", t.shape)              # (2, 3)
print("  -> filas   :", t.shape[0])
print("  -> columnas:", t.shape[1])
print("dtype (tipo)            :", t.dtype)              # float32
print("size (total elementos)  :", tf.size(t).numpy())   # 6
print("device (dispositivo)    :", t.device)
````

## 9. Cómo crear tensores

`tf.constant()` es la forma básica, pero hay atajos para los casos habituales:

````python
# --- Desde valores explícitos ---
a = tf.constant([[10.0, 20.0], [30.0, 40.0], [50.0, 60.0]])   # matriz 3x2
b = tf.constant(42)                                            # escalar

# --- Rellenos ---
ceros  = tf.zeros([2, 3])            # todo 0
unos   = tf.ones([2, 3])             # todo 1
lleno  = tf.fill([2, 3], 7.0)        # todo 7
ident  = tf.eye(3)                   # matriz identidad 3x3

# --- Secuencias y aleatorios ---
rango  = tf.range(0, 10, 2)                              # 0,2,4,6,8
azar   = tf.random.normal([2, 3], mean=0.0, stddev=1.0)  # distribución normal

# --- Desde NumPy ---
desde_np = tf.constant(np.array([[1.0, 2.0], [3.0, 4.0]]))

print("a =\n", a.numpy(), "\n")
print("b =", b.numpy())
print("rango =", rango.numpy())
print("identidad =\n", ident.numpy())
print("azar =\n", azar.numpy())
print("desde numpy -> dtype:", desde_np.dtype, "  (¡ojo: float64, no float32!)")
````

## 10. Indexado y slicing

Funciona **igual que en NumPy**. La sintaxis es `t[filas, columnas]`:

| Quiero... | Escribo |
|---|---|
| La primera fila | `t[0]` o `t[0, :]` |
| La última fila | `t[-1]` |
| La primera columna | `t[:, 0]` |
| La **última** columna | `t[:, -1]` |
| Un elemento suelto | `t[1, 2]` |
| Las 2 primeras filas | `t[:2]` |
| Una submatriz | `t[0:2, 1:3]` |

> 🔑 **La regla que resuelve todo:** los dos puntos `:` significan *"todos los de este eje"*. Entonces `t[:, -1]` se lee: *"todas las filas, última columna"*.

````python
t = tf.constant([[10., 20., 30.],
                 [40., 50., 60.]])
print("t =\n", t.numpy(), "\n")

print("t[0]      (primera fila)   ->", t[0].numpy())
print("t[-1]     (última fila)    ->", t[-1].numpy())
print("t[:, 0]   (primera columna)->", t[:, 0].numpy())
print("t[:, -1]  (última columna) ->", t[:, -1].numpy())
print("t[1, 2]   (un elemento)    ->", t[1, 2].numpy())
print()
print("t[:, 1:]  (de la col. 1 en adelante) ->\n", t[:, 1:].numpy())

# Mantener la dimensión: con tf.newaxis la columna sigue siendo una columna
print("\nt[:, -1]             shape:", t[:, -1].shape, " (se aplanó a vector)")
print("t[:, -1, tf.newaxis] shape:", t[:, -1, tf.newaxis].shape, " (sigue siendo columna)")
````

## 11. Operaciones

### 11.1 Elemento a elemento (*element-wise*)

Las operaciones aritméticas comunes se aplican **posición por posición**:

````python
t = tf.constant([[10., 20., 30.],
                 [40., 50., 60.]])

print("t + 5   (suma un escalar a cada elemento):\n", (t + 5.0).numpy(), "\n")
print("t + t   (suma dos tensores posición a posición):\n", (t + t).numpy(), "\n")
print("t * 2   :\n", (t * 2).numpy(), "\n")
print("t ** 2  :\n", (t ** 2).numpy())
````

````python
# Funciones matemáticas: se aplican a cada elemento
a = tf.constant([4.0, 9.0, 16.0, 25.0])

print("a           :", a.numpy())
print("tf.sqrt(a)  :", tf.sqrt(a).numpy())        # raíz cuadrada
print("tf.square(a):", tf.square(a).numpy())      # al cuadrado
print("tf.exp(a)   :", tf.exp(a).numpy())
print("tf.abs(-a)  :", tf.abs(-a).numpy())

# Cada operador tiene su equivalente como función:
print("\nt + t  ==  tf.add(t, t)      ->", bool(tf.reduce_all(tf.add(t, t) == t + t)))
print("t * 2  ==  tf.multiply(t, 2) ->", bool(tf.reduce_all(tf.multiply(t, 2) == t * 2)))
````

### 11.2 ⚠️ `*` NO es multiplicación de matrices

Este es **el error clásico** de todo el que arranca:

- `A * B` → multiplica **elemento a elemento** (las shapes deben coincidir)
- `A @ B` → **producto matricial** (equivale a `tf.matmul(A, B)`)

Para el producto matricial, las dimensiones internas tienen que coincidir: `(2,3) @ (3,2)` → `(2,2)`.

````python
A = tf.constant([[1., 2., 3.],
                 [4., 5., 6.]])          # shape (2, 3)
B = tf.transpose(A)                      # shape (3, 2)

print("A shape:", A.shape, " B shape:", B.shape, "\n")

print("A @ B  (producto matricial) -> shape", (A @ B).shape)
print((A @ B).numpy(), "\n")

print("A * A  (elemento a elemento) -> shape", (A * A).shape)
print((A * A).numpy())
````

### 11.3 Broadcasting

Cuando las shapes no son idénticas, TensorFlow intenta **estirar** la más chica para que encajen. Es lo que permite escribir `t + 5` sin crear una matriz llena de cincos.

**La regla:** se comparan las dimensiones de derecha a izquierda. Son compatibles si **son iguales** o si **alguna vale 1**.

```
   (2, 3)  +  (3,)   ->  ✅  el vector se repite en cada fila
   (2, 3)  +  (2, 1) ->  ✅  la columna se repite en cada columna
   (2, 3)  +  (2,)   ->  ❌  3 y 2 no son compatibles
```

````python
t = tf.constant([[10., 20., 30.],
                 [40., 50., 60.]])          # (2, 3)

fila    = tf.constant([1., 2., 3.])          # (3,)   -> se repite en cada fila
columna = tf.constant([[100.], [200.]])      # (2, 1) -> se repite en cada columna

print("t + fila    (broadcast por filas):\n", (t + fila).numpy(), "\n")
print("t + columna (broadcast por columnas):\n", (t + columna).numpy())

# El caso que NO funciona:
try:
    t + tf.constant([1., 2.])                # (2,) contra (2,3) -> incompatible
except Exception as e:
    print("\n❌ t + [1., 2.] falla:", type(e).__name__)
````

### 11.4 Reducciones y el parámetro `axis`

Las funciones `reduce_*` **colapsan** un eje. El parámetro `axis` decide cuál:

- `axis=None` (por defecto) → reduce **todo** a un escalar
- `axis=0` → colapsa las **filas** ⇒ resultado *por columna*
- `axis=1` → colapsa las **columnas** ⇒ resultado *por fila*

> 🔑 **Truco para no confundirse:** `axis` indica el eje que **desaparece**.

````python
t = tf.constant([[10., 20., 30.],
                 [40., 50., 60.]])

print("t =\n", t.numpy(), "\n")
print("reduce_sum(t)          ->", tf.reduce_sum(t).numpy(), "  (todo)")
print("reduce_sum(t, axis=0)  ->", tf.reduce_sum(t, axis=0).numpy(), "  (una suma por columna)")
print("reduce_sum(t, axis=1)  ->", tf.reduce_sum(t, axis=1).numpy(), "  (una suma por fila)")
print()
print("reduce_mean(t)         ->", tf.reduce_mean(t).numpy())
print("reduce_max(t, axis=1)  ->", tf.reduce_max(t, axis=1).numpy())
print("argmax(t, axis=1)      ->", tf.argmax(t, axis=1).numpy(), "  (posición del máximo por fila)")
````

### 11.5 Cambiar la forma

`reshape` reordena los mismos datos en otra forma; `transpose` intercambia los ejes. Con `-1` en `reshape`, TensorFlow calcula solo esa dimensión.

````python
t = tf.constant([[10., 20., 30.],
                 [40., 50., 60.]])          # (2, 3)

print("original      (2,3):\n", t.numpy(), "\n")
print("reshape (3,2) :\n", tf.reshape(t, [3, 2]).numpy(), "\n")
print("reshape (6,)  :", tf.reshape(t, [-1]).numpy(), "   <- aplanar (muy usado antes de una capa densa)")
print("\ntranspose     :\n", tf.transpose(t).numpy())
````

## 12. Tipos de datos y `tf.cast`

> ⚠️ **¡Cuidado!** TensorFlow **no realiza conversión implícita de tipos** (*casting implícito*).

Si operás dos tensores de distinto `dtype`, **se produce una excepción**. Python te dejaría hacer `1 + 1.0`; TensorFlow no.

**Por qué es así (y no es un capricho):** en la GPU cada operación está compilada para un tipo concreto. Una conversión automática y silenciosa podría duplicar el uso de memoria o degradar la precisión sin que te enteres — en un modelo con millones de parámetros, eso importa. TensorFlow prefiere fallar ruidosamente.

**Los dtypes que vas a ver:**

| Dtype | Uso |
|---|---|
| `tf.float32` | **El estándar en Deep Learning.** Es el default para decimales de Python |
| `tf.float64` | Doble precisión. Es el default de NumPy — de ahí vienen muchos choques |
| `tf.int32` | Default para enteros de Python |
| `tf.string` | Texto (se guarda como **bytes**) |
| `tf.bool` | Booleanos |

**La solución:** `tf.cast(tensor, tipo_destino)`.

````python
a = tf.constant([1, 2, 3])          # int32   (enteros sin punto decimal)
b = tf.constant([1.0, 2.0, 3.0])    # float32 (con punto decimal)

print("a:", a.numpy(), "-> dtype:", a.dtype)
print("b:", b.numpy(), "-> dtype:", b.dtype)

# 1) El error
try:
    print(a + b)
except Exception as e:
    print("\n❌ a + b falla ->", type(e).__name__)
    print("   ", str(e).split("\n")[0][:130])

# 2) La solución
a_float = tf.cast(a, tf.float32)
print("\n✅ Con tf.cast(a, tf.float32):", a_float.dtype)
print("   a_float + b =", (a_float + b).numpy())
````

````python
# Otras formas de fijar el tipo desde el principio
print("Forzando el dtype al crear:", tf.constant([1, 2, 3], dtype=tf.float32).dtype)

# Ojo con NumPy: su default es float64 y choca con el float32 de TF
np_arr = np.array([1.0, 2.0, 3.0])          # float64
print("\nDesde NumPy sin cuidado ->", tf.constant(np_arr).dtype, " ⚠️")
print("Corregido               ->", tf.constant(np_arr, dtype=tf.float32).dtype, " ✅")

# El cast a entero TRUNCA (no redondea)
print("\ncast(3.9 -> int32):", tf.cast(tf.constant([3.9, -2.7]), tf.int32).numpy(), " <- trunca hacia cero")
print("tf.round([3.9,-2.7]):", tf.round(tf.constant([3.9, -2.7])).numpy())
````

## 13. `tf.constant` vs. `tf.Variable`

| | `tf.constant` 🔒 | `tf.Variable` 📦 |
|---|---|---|
| Mutabilidad | **Inmutable** | **Mutable** |
| Modificar | ❌ `t[0] = 4` da error | ✅ `v.assign(...)` |
| Para qué sirve | Datos de entrada, constantes, resultados intermedios | **Pesos y bias**, que cambian durante el entrenamiento |

**Por qué existe esta distinción:** entrenar una red neuronal *es* ir modificando los pesos paso a paso. Si todo fuera inmutable, cada actualización obligaría a recrear el tensor entero. `tf.Variable` es el contenedor pensado para ese estado que cambia.

**Métodos de `tf.Variable`:**
- `v.value()` → devuelve el tensor con el valor actual
- `v.assign(nuevo)` → reemplaza **todo** el contenido
- `v[i].assign(x)` → modifica **un elemento**
- `v.assign_add(x)` / `v.assign_sub(x)` → suma o resta en el lugar

> ⚠️ La shape y el dtype **no** se pueden cambiar con `assign`: el valor nuevo tiene que tener la misma forma y tipo.

````python
# 1) La inmutabilidad de tf.constant
t = tf.constant([10.0, 20.0, 30.0])
try:
    t[0] = 99.0
except Exception as e:
    print("❌ tf.constant NO se puede modificar ->", type(e).__name__)
    print("  ", e)
````

````python
# 2) tf.Variable sí es mutable
v = tf.Variable([100.0, 200.0, 300.0])

print("Variable original      :", v.value().numpy())

v[1].assign(250.0)                      # modifica SOLO el índice 1
print("Tras v[1].assign(250.0):", v.value().numpy())

v.assign([1.0, 2.0, 3.0])               # reemplaza todo
print("Tras v.assign([1,2,3]) :", v.value().numpy())

v.assign_add([10.0, 10.0, 10.0])        # suma en el lugar
print("Tras v.assign_add(+10) :", v.value().numpy())

print("\nTipo del objeto:", type(v).__name__)
print("Atributos: shape =", v.shape, "| dtype =", v.dtype.name, "| trainable =", v.trainable)
````

## 14. Tensores de texto

TensorFlow también maneja cadenas (`tf.string`). El detalle importante: **internamente se guardan como bytes**, por eso al imprimirlas aparece el prefijo `b'...'`.

Además, un tensor de strings **no tiene una dimensión por carácter**: cada cadena completa es *un solo elemento*. Un tensor de 3 palabras tiene shape `(3,)`, sin importar cuán largas sean.

````python
s1 = tf.constant(b"machine learning")                          # ya en bytes
s2 = tf.constant("inteligencia artificial")                    # str -> se convierte solo
s3 = tf.constant(["python", "tensorflow", "data science"])     # vector de strings

print("s1:", s1, "\n")
print("s2:", s2, "\n")
print("s3:", s3, "\n")

print("shape de s3:", s3.shape, " <- 3 elementos (no importa el largo de cada palabra)")
print("dtype       :", s3.dtype)

# Volver a texto legible
print("\nDecodificado:", [x.decode("utf-8") for x in s3.numpy()])

# Algunas operaciones sobre strings
print("\nMayúsculas :", tf.strings.upper(s3).numpy())
print("Longitudes  :", tf.strings.length(s3).numpy())
print("Split de s2 :", tf.strings.split(s2).numpy())
````

## 15. Puente con NumPy

Los dos mundos se conectan sin fricción — y en la práctica vas a cruzar esa frontera todo el tiempo (los datos llegan en NumPy o pandas, el modelo los quiere en tensores).

````python
t = tf.constant([[1., 2.], [3., 4.]])

# TensorFlow -> NumPy
arr = t.numpy()
print("t.numpy() ->", type(arr).__name__, "\n", arr, "\n")

# NumPy -> TensorFlow
back = tf.constant(arr)
print("tf.constant(arr) ->", type(back).__name__, "| dtype:", back.dtype, "\n")

# Muchas funciones de NumPy aceptan tensores directamente
print("np.sum(t)  ->", np.sum(t))
print("np.mean(t) ->", np.mean(t))
````

## 16. Los 5 errores más frecuentes

| # | Error | Causa | Solución |
|---|---|---|---|
| 1 | `InvalidArgumentError: cannot compute AddV2...` | Sumaste `int32` con `float32` | `tf.cast(a, tf.float32)` |
| 2 | `TypeError: ...object does not support item assignment` | Intentaste `t[0] = x` en un `tf.constant` | Usá `tf.Variable` + `.assign()` |
| 3 | `InvalidArgumentError: Incompatible shapes` | Las shapes no son compatibles ni por broadcasting | Revisá `t.shape`; usá `reshape` o `tf.newaxis` |
| 4 | `Matrix size-incompatible` | Confundiste `*` con `@` | Para producto matricial usá `@`; las dims internas deben coincidir |
| 5 | El resultado da `float64` inesperado | Los datos vinieron de NumPy | `tf.constant(arr, dtype=tf.float32)` |

> **Método general para depurar:** cuando algo falla, imprimí **siempre** `t.shape` y `t.dtype` de los tensores involucrados. El 90% de los errores de TensorFlow son de forma o de tipo.
