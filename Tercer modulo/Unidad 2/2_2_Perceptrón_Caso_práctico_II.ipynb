## Dataset

La base de datos del MNIST (base de datos modificada del Instituto Nacional de Normas y Tecnología)
La base de datos del MNIST contiene 60.000 imágenes de entrenamiento y 10.000 imágenes de prueba.

## Enunciado

En este ejercicio práctico vamos a utilizar el Perceptrón para solucionar un problema más complejo que el anterior. En este caso, nuestro objetivo será clasificar un conjunto de imágener en base al número que se muestra en las mismas.

## 1. Lectura del conjunto de datos

```python
# Importamos el conjunto de datos
from tensorflow.keras.datasets import mnist

(X_train, y_train), (X_test, y_test) = mnist.load_data()   #ver shape
```

## 2. Visualización del conjunto de datos

```python
import numpy as np
import matplotlib.pyplot as plt

plt.figure(figsize=(28, 28))

for index, img in zip(range(1, 9), X_train[:8]):
    plt.subplot(1, 8, index)
    plt.imshow(img, cmap='gray')
    plt.title('Ejemplo: ' + str(index))
    plt.axis('off')
plt.show()
```

## 4. Entrenamiento del algoritmo

```python
from sklearn.linear_model import Perceptron

# Reshape X_train and X_test to 2D arrays
#X_train tiene una forma (60000, 28, 28) y debo aplanar a 2 dimenciones para el precptron uso reshaped
X_train_reshaped = X_train.reshape(X_train.shape[0], -1)
X_test_reshaped = X_test.reshape(X_test.shape[0], -1)

modelo = Perceptron(max_iter=2000, random_state=40, n_jobs=-1)

modelo.fit(X_train_reshaped, y_train)
```

```python
X_train_reshaped.shape
```

```python
# Número de parámetros pesos que forman el modelo
modelo.coef_.shape
```

```python
# Parámetros bias/intercept
modelo.intercept_
```

## 5. Predicción con el conjunto de pruebas

```python
# Realizamos la predicción con el conjunto de datos de prueba
y_pred = modelo.predict(X_test_reshaped)
```

```python
# Mostramos el f1_score resultante de la clasificación
from sklearn.metrics import f1_score

f1_score(y_test, y_pred, average="weighted")
```

## 6. Mostrando las imagenes mal clasificadas

```python
import numpy as np
import matplotlib.pyplot as plt

index = 0
index_errors = []

for label, predict in zip(y_test, y_pred):
    if label != predict:
        index_errors.append(index)
    index += 1
```

```python
# Graficar algunas imágenes mal clasificadas (ejemplo: índices 8 al 15)
plt.figure(figsize=(20, 4))

for i, img_index in zip(range(1, 9), index_errors[8:16]):
    plt.subplot(1, 8, i)
    # TODO: Redimensionar la imagen a 28x28 e imprimirla con su valor real y predicho
    plt.imshow(np.reshape(X_test[img_index], (28,28)), cmap=plt.cm.gray)
    plt.title('Orig:' + str(y_test[img_index]) + ' Pred:' + str(y_pred[img_index]))
plt.show()
```
