# Primera red neuronal: Celsius → Fahrenheit (regresión)

En este ejemplo construiremos una red neuronal muy sencilla para aprender la relación entre **grados Celsius y Fahrenheit**.

La fórmula matemática es:

$$F = C * 1.8 + 32$$

La red neuronal intentará aprender esta relación a partir de ejemplos.

```python
import tensorflow as tf
import numpy as np

print('TensorFlow:', tf.__version__)
```

## 1. Datos de entrenamiento

Le damos a la red algunos valores conocidos de Celsius y su equivalente en Fahrenheit.

```python
celsius = np.array([-40, -10, 0, 8, 15, 22, 38, 50], dtype=float)
fahrenheit = np.array([-40, 14, 32, 46.4, 59, 71.6, 100.4, 122], dtype=float)

print('Celsius:', celsius)
print('Fahrenheit:', fahrenheit)
```

## 2. Crear la red neuronal

Utilizamos una única neurona (`Dense(1)`).

La neurona deberá aprender dos parámetros que se parecen a los de la fórmula matemática: el **peso** y el **sesgo (bias)**.

```python
capa = tf.keras.layers.Dense(units=1, input_shape=[1])

modelo = tf.keras.Sequential([capa])

```

```python
modelo.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.1),
    loss='mean_squared_error'
)

modelo.summary()   # 1 entrada X 1 neurona + 1 bias = 2 parametros
```

## 3. Entrenar la red

Durante el entrenamiento, la red compara sus predicciones con los valores reales y ajusta sus pesos para reducir el error.

```python
history = modelo.fit(
    celsius,
    fahrenheit,
    epochs=500,
    verbose=0
)
print('history:', history.history.keys())
print('Entrenamiento terminado')
print('Error final:', history.history['loss'][-1])
```

```python
len(modelo.layers)
```

## 4. Observar lo que aprendió la neurona

Para una neurona `Dense(1)`, la salida tiene la forma:

$$y = x \times peso + bias$$

```python
peso, bias = modelo.layers[0].get_weights()
print('Peso:', peso)
print('Bias:', bias)

print('Peso aprendido:', peso[0][0])
print('Bias aprendido:', bias[0])
print('\nValores teóricos:')
print('Peso = ', 1.8)
print('Bias = 32')
```

## 5. Hacer una predicción

Ahora le preguntamos a la red neuronal cuántos grados Fahrenheit corresponden a **100 °C**.

```python
celsius_nuevo = np.array([100.0])
prediccion = modelo.predict(celsius_nuevo, verbose=0)

print(f'100 °C → {prediccion[0][0]:.2f} °F')
print('Valor correcto → 212 °F')
```

## 6. Probar otros valores


```python
valores = np.array([0, 10, 20, 30, 40, 50, 100], dtype=float)
predicciones = modelo.predict(valores, verbose=0).flatten()

for c, f in zip(valores, predicciones):
    print(f'{c:6.1f} °C → {f:7.2f} °F')
```

## Conclusión

Esta es una red neuronal extremadamente simple: **una entrada, una neurona y una salida**.

Aunque la relación Celsius-Fahrenheit puede resolverse con una fórmula, este ejemplo permite observar el principio fundamental del aprendizaje automático:

**datos → predicción → cálculo del error → ajuste de pesos → nueva predicción**.
